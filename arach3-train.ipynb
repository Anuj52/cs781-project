```python
# ============================================================
# CS781 Phase 1 — Architecture 3
# Sliding-Window Attention (SWA) — Full Training Script
# ============================================================

import os
import time
import math
import json
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F

# ============================================================
# 0. MODEL DEFINITION
# ============================================================

def sinusoidal_pos_emb(T, d_model, device):
    """
    Fixed sinusoidal position encoding.
    Shape: (1, T, d_model). No learned parameters.
    """
    pos = torch.arange(T, device=device, dtype=torch.float32).unsqueeze(1)

    div = torch.exp(
        torch.arange(0, d_model, 2, device=device, dtype=torch.float32)
        * (-math.log(10000.0) / d_model)
    )

    pe = torch.zeros(T, d_model, device=device, dtype=torch.float32)
    pe[:, 0::2] = torch.sin(pos * div)
    pe[:, 1::2] = torch.cos(pos * div)

    return pe.unsqueeze(0)  # (1, T, d_model)


class SlidingWindowCausalSelfAttention(nn.Module):
    def __init__(self, d_model, n_heads, block_size, window_size):
        super().__init__()

        assert d_model % n_heads == 0

        self.d_model = d_model
        self.n_heads = n_heads
        self.head_dim = d_model // n_heads
        self.window_size = window_size

        self.q_proj = nn.Linear(d_model, d_model)
        self.k_proj = nn.Linear(d_model, d_model)
        self.v_proj = nn.Linear(d_model, d_model)
        self.out_proj = nn.Linear(d_model, d_model)

        # Mask rule:
        # token i may attend to token j iff j <= i and i - j < W.
        #
        # For torch.nn.functional.scaled_dot_product_attention,
        # a boolean mask means True = allowed.
        idx = torch.arange(block_size)

        causal = idx[None, :] <= idx[:, None]          # j <= i
        window = (idx[:, None] - idx[None, :]) < window_size  # i - j < W

        mask = causal & window

        self.register_buffer(
            "swa_mask",
            mask.view(1, 1, block_size, block_size),
            persistent=False,
        )

    def forward(self, x):
        B, T, C = x.shape

        q = self.q_proj(x)
        k = self.k_proj(x)
        v = self.v_proj(x)

        q = q.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        k = k.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        v = v.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)

        # Slice the precomputed mask for shorter inputs.
        attn_mask = self.swa_mask[:, :, :T, :T]

        # Do NOT use is_causal=True together with an explicit mask.
        # The banded mask already includes causality.
        y = F.scaled_dot_product_attention(
            q,
            k,
            v,
            attn_mask=attn_mask,
            dropout_p=0.0,
            is_causal=False,
        )

        y = y.transpose(1, 2).contiguous().view(B, T, C)

        return self.out_proj(y)


class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()

        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)

    def forward(self, x):
        return self.fc2(F.gelu(self.fc1(x)))


class TransformerBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, block_size, window_size):
        super().__init__()

        self.norm1 = nn.LayerNorm(d_model)

        self.attn = SlidingWindowCausalSelfAttention(
            d_model=d_model,
            n_heads=n_heads,
            block_size=block_size,
            window_size=window_size,
        )

        self.norm2 = nn.LayerNorm(d_model)

        self.mlp = FeedForward(d_model, d_ff)

    def forward(self, x):
        # Pre-LN residual block.
        x = x + self.attn(self.norm1(x))
        x = x + self.mlp(self.norm2(x))
        return x


class LegalDecoderLM(nn.Module):
    def __init__(
        self,
        vocab_size,
        block_size,
        d_model,
        n_layers,
        n_heads,
        d_ff,
        window_size,
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.block_size = block_size
        self.d_model = d_model
        self.n_layers = n_layers

        self.token_emb = nn.Embedding(vocab_size, d_model)

        # Scale token embeddings by sqrt(d_model) so they are comparable
        # in magnitude to the sinusoidal position table.
        self.embed_scale = math.sqrt(d_model)

        # Precompute sinusoidal positions once. Not a parameter.
        pos = sinusoidal_pos_emb(block_size, d_model, device="cpu")
        self.register_buffer("pos_emb", pos, persistent=False)

        self.blocks = nn.ModuleList([
            TransformerBlock(
                d_model=d_model,
                n_heads=n_heads,
                d_ff=d_ff,
                block_size=block_size,
                window_size=window_size,
            )
            for _ in range(n_layers)
        ])

        self.final_norm = nn.LayerNorm(d_model)

        self.lm_head = nn.Linear(d_model, vocab_size, bias=False)

        # Tied embedding and output head.
        self.lm_head.weight = self.token_emb.weight

        self.apply(self._init_weights)

        # GPT-style residual scaling for deep pre-LN stacks.
        for block in self.blocks:
            block.attn.out_proj.weight.data.mul_(1.0 / math.sqrt(2 * self.n_layers))
            block.mlp.fc2.weight.data.mul_(1.0 / math.sqrt(2 * self.n_layers))

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                nn.init.zeros_(module.bias)

        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

        elif isinstance(module, nn.LayerNorm):
            nn.init.ones_(module.weight)
            nn.init.zeros_(module.bias)

    def forward(self, x):
        B, T = x.shape

        assert T <= self.block_size, (
            f"Sequence length {T} exceeds block_size {self.block_size}"
        )

        tok = self.token_emb(x) * self.embed_scale

        pos = self.pos_emb[:, :T, :].to(tok.dtype)

        h = tok + pos

        for block in self.blocks:
            h = block(h)

        h = self.final_norm(h)

        return self.lm_head(h)


def build_model(vocab_size, block_size, window_size=256):
    """
    Architecture 3 size:
      d_model = 1024
      n_layers = 16
      n_heads = 16
      d_ff = 4096
      W = 256 by default

    Parameter count with vocab_size=40000:
      242,501,632 ≈ 242.5M
    """
    d_model = 1024
    n_layers = 16
    n_heads = 16
    d_ff = 4096

    return LegalDecoderLM(
        vocab_size=vocab_size,
        block_size=block_size,
        d_model=d_model,
        n_layers=n_layers,
        n_heads=n_heads,
        d_ff=d_ff,
        window_size=window_size,
    )


# ============================================================
# 1. PATHS / CONFIG
# ============================================================

TOKD = "/kaggle/input/datasets/achiver7/phase1-tokens-v40k-clean1/tokens"

# Use a unique run name for this experiment
RUN = "/kaggle/working/run_arch3_swa_v2"

# True : start from random weights and REFUSE to run if this folder already holds a checkpoint
#         (prevents a silent resume). False: resume from latest.pt if it exists.
FRESH_START = True

V = 40000
TIME_LIMIT_H = 8.25
dev = "cuda"

CFG = dict(
    BLOCK=1024,
    MICRO_BATCH=2,      # speed only: 2 x 8 gives the same 16,384 tokens per optimizer step as 1 x 16
    GRAD_ACCUM=8,

    LR=3e-4,
    MIN_LR_FRAC=0.1,
    WD=0.1,
    WARMUP=500,

    STEPS=10000,
    CLIP=1.0,
    EVAL_EVERY=500,
    SEED=42,

    # Sliding window size
    W=256,

    MIX=dict(
        acts=0.273,
        sc=0.469,
        tri=0.258
    )
)

# ============================================================
# 2. BASIC CHECKS
# ============================================================

assert torch.cuda.is_available(), "CUDA GPU required"
assert os.path.exists(TOKD), f"Missing: {TOKD}"
assert os.path.exists(f"{TOKD}/tokenizer.json")
assert callable(build_model), "build_model() not found"

assert abs(sum(CFG["MIX"].values()) - 1.0) < 1e-6

BLOCK = CFG["BLOCK"]
MB = CFG["MICRO_BATCH"]
GA = CFG["GRAD_ACCUM"]
W = CFG["W"]

assert MB * GA * BLOCK == 16384, "tokens per optimizer step changed; the run was started with 16,384"

np.random.seed(CFG["SEED"])
torch.manual_seed(CFG["SEED"])
torch.cuda.manual_seed_all(CFG["SEED"])

os.makedirs(RUN, exist_ok=True)

print("=" * 70)
print("PHASE 1 — ARCHITECTURE 3 (SWA) — REAL TRAINING")
print("=" * 70)
print("GPU:", torch.cuda.get_device_name(0))
print(
    "VRAM:",
    round(
        torch.cuda.get_device_properties(0).total_memory / 1e9,
        2
    ),
    "GB"
)
print("BLOCK:", BLOCK)
print("WINDOW W:", W)
print("MICRO_BATCH:", MB)
print("GRAD_ACCUM:", GA)
print(
    "TOKENS / OPTIMIZER STEP:",
    MB * GA * BLOCK
)
print("TOTAL STEPS:", CFG["STEPS"])
print("TIME LIMIT:", TIME_LIMIT_H, "hours")
print()

# ============================================================
# 3. TOKEN DATA
# ============================================================

def mm(name):
    path = f"{TOKD}/{name}.bin"
    assert os.path.exists(path), f"Missing token file: {path}"
    return np.memmap(
        path,
        dtype=np.uint16,
        mode="r"
    )

data = {
    "acts": mm("acts_train"),
    "sc": mm("sc_train"),
    "tri": mm("tribunals_train_f"),
}

valbins = {
    "acts_val": mm("acts_val"),
    "sc_val": mm("sc_val"),
    "tribunals_val_f": mm("tribunals_val_f"),
}

keys = list(CFG["MIX"].keys())
probs = list(CFG["MIX"].values())

print("TRAIN TOKENS")
for name, d in data.items():
    print(f"{name:5s}: {len(d):,}")

print()

# ============================================================
# 4. BATCHING
# ============================================================

def get_batch():
    xs = []
    ys = []

    for _ in range(MB):

        source = keys[
            np.random.choice(
                len(keys),
                p=probs
            )
        ]

        d = data[source]

        i = np.random.randint(
            0,
            len(d) - BLOCK - 1
        )

        chunk = d[
            i:i + BLOCK + 1
        ].astype(np.int64)

        xs.append(chunk[:-1])
        ys.append(chunk[1:])

    x = torch.from_numpy(
        np.stack(xs)
    ).to(dev)

    y = torch.from_numpy(
        np.stack(ys)
    ).to(dev)

    return x, y

# ============================================================
# 5. FRESH MODEL
# ============================================================

print("Building Architecture 3 model...")

model = build_model(V, BLOCK, W).to(dev)

n_params = sum(
    p.numel()
    for p in model.parameters()
)

print(
    f"PARAMETERS: {n_params:,}"
)
print(
    f"PARAMETERS: {n_params / 1e6:.2f}M"
)

assert 200e6 <= n_params <= 250e6, (
    f"Invalid parameter count: {n_params/1e6:.2f}M"
)

# ============================================================
# 6. OPTIMIZER + AMP
# ============================================================

try:
    opt = torch.optim.AdamW(
        model.parameters(),
        lr=CFG["LR"],
        weight_decay=CFG["WD"],
        fused=True
    )
    print("AdamW: fused")
except Exception as e:
    print("fused AdamW unavailable, using the default implementation:", e)
    opt = torch.optim.AdamW(
        model.parameters(),
        lr=CFG["LR"],
        weight_decay=CFG["WD"]
    )

scaler = torch.amp.GradScaler("cuda")

# ============================================================
# 7. LEARNING RATE
# ============================================================

def lr_at(step):

    if step < CFG["WARMUP"]:
        return (
            CFG["LR"]
            * (step + 1)
            / CFG["WARMUP"]
        )

    p = (
        step - CFG["WARMUP"]
    ) / max(
        1,
        CFG["STEPS"] - CFG["WARMUP"]
    )

    p = min(
        max(p, 0.0),
        1.0
    )

    return CFG["LR"] * (
        CFG["MIN_LR_FRAC"]
        + (
            1 - CFG["MIN_LR_FRAC"]
        )
        * 0.5
        * (
            1 + math.cos(
                math.pi * p
            )
        )
    )

# ============================================================
# 8. FIXED VALIDATION WINDOWS
# ============================================================

rs = np.random.RandomState(123)

val_windows = {}

for name, d in valbins.items():

    starts = rs.randint(
        0,
        len(d) - BLOCK - 1,
        200
    )

    val_windows[name] = np.stack([
        d[
            i:i + BLOCK + 1
        ].astype(np.int64)
        for i in starts
    ])

print("Validation windows ready.")
print()

# ============================================================
# 9. TRAIN STEP
# ============================================================

def train_step(lr):

    for g in opt.param_groups:
        g["lr"] = lr

    total_loss = torch.zeros((), device=dev)

    for _ in range(GA):

        x, y = get_batch()

        with torch.autocast(
            "cuda",
            dtype=torch.float16
        ):
            logits = model(x)

        loss = F.cross_entropy(
            logits.float().reshape(-1, V),
            y.reshape(-1)
        )

        loss = loss / GA

        scaler.scale(loss).backward()

        total_loss += loss.detach()

        del x, y, logits, loss

    scaler.unscale_(opt)

    grad_norm = nn.utils.clip_grad_norm_(
        model.parameters(),
        CFG["CLIP"]
    )

    scaler.step(opt)
    scaler.update()

    opt.zero_grad(
        set_to_none=True
    )

    return (
        float(total_loss),
        float(grad_norm)
    )

# ============================================================
# 10. VALIDATION
# ============================================================

@torch.no_grad()
def val_loss(bs=1):

    model.eval()

    results = {}

    for name, windows in val_windows.items():

        total = 0.0
        count = 0

        for i in range(
            0,
            len(windows),
            bs
        ):

            b = torch.from_numpy(
                windows[i:i + bs]
            ).to(dev)

            with torch.autocast(
                "cuda",
                dtype=torch.float16
            ):
                logits = model(
                    b[:, :-1]
                )

            total += F.cross_entropy(
                logits.float().reshape(-1, V),
                b[:, 1:].reshape(-1),
                reduction="sum"
            ).item()

            count += b[:, 1:].numel()

            del b, logits

        results[name] = total / count

    model.train()

    return results

# ============================================================
# 11. CHECKPOINT
# ============================================================

def save_ckpt(path, step, best, full=True):
    """full=True : weights + optimizer + scaler (needed to resume).
    full=False: weights only (enough to generate, about a third of the size).
    Written to a temporary file and then renamed, so an interrupted write cannot corrupt an existing checkpoint."""
    ck = {
        "model": model.state_dict(),
        "step": step,
        "best": best,
        "cfg": CFG,
        "n_params": n_params,
    }
    if full:
        ck["opt"] = opt.state_dict()
        ck["scaler"] = scaler.state_dict()

    tmp = path + ".tmp"
    torch.save(ck, tmp)
    os.replace(tmp, path)

# ============================================================
# 12. REAL TRAINING
# ============================================================

def run_training():

    step0 = 0
    best = float("inf")

    latest_path = f"{RUN}/latest.pt"

    if FRESH_START:
        assert not os.path.exists(latest_path), (
            f"{latest_path} already exists. Use a new RUN folder for a fresh start, or set FRESH_START = False to resume."
        )

    # Resume if this run already contains a checkpoint
    if os.path.exists(latest_path):

        print("=" * 70)
        print("CHECKPOINT FOUND — RESUMING")
        print("=" * 70)

        ck = torch.load(
            latest_path,
            map_location=dev,
            weights_only=False
        )

        model.load_state_dict(
            ck["model"]
        )

        opt.load_state_dict(
            ck["opt"]
        )

        scaler.load_state_dict(
            ck["scaler"]
        )

        step0 = ck["step"]
        best = ck["best"]

        del ck
        torch.cuda.empty_cache()

        # Re-seed the data sampler with a step-dependent seed. With the original seed a resumed run would
        # draw the same random windows it already used from step 0.
        np.random.seed(CFG["SEED"] + step0)
        torch.manual_seed(CFG["SEED"] + step0)

        print("Resume step:", step0)
        print("Best validation:", best)
        print()

    else:

        print("=" * 70)
        print("STARTING FROM FRESH RANDOM WEIGHTS")
        print("=" * 70)
        print()

    model.train()

    run_start = time.time()
    last_log_time = run_start
    last_log_step = step0

    history = []

    for step in range(
        step0,
        CFG["STEPS"]
    ):

        lr = lr_at(step)

        loss, grad_norm = train_step(lr)

        history.append(loss)

        elapsed = (
            time.time()
            - run_start
        )

        timeout = (
            elapsed >= TIME_LIMIT_H * 3600
        )

        evaluate = (
            (step + 1) % CFG["EVAL_EVERY"] == 0
            or (step + 1) == CFG["STEPS"]
            or timeout
        )

        if evaluate:

            val = val_loss(bs=1)

            val_mean = float(
                np.mean(
                    list(val.values())
                )
            )

            now = time.time()

            dt = max(
                now - last_log_time,
                1e-9
            )

            delta_steps = (
                step + 1
                - last_log_step
            )

            tok_per_sec = (
                delta_steps
                * MB
                * GA
                * BLOCK
                / dt
            )

            tokens_seen = (
                (step + 1)
                * MB
                * GA
                * BLOCK
            )

            print()
            print(
                "-" * 70
            )
            print(
                f"STEP {step+1:,}/{CFG['STEPS']:,}"
            )
            print(
                f"TOKENS: {tokens_seen/1e6:,.2f}M"
            )
            print(
                f"TRAIN LOSS: {np.mean(history[-50:]):.4f}"
            )

            for k, v in val.items():
                print(
                    f"{k}: {v:.4f}"
                )

            print(
                f"VAL MEAN: {val_mean:.4f}"
            )
            print(
                f"LR: {lr:.6e}"
            )
            print(
                f"GRAD NORM: {grad_norm:.4f}"
            )
            print(
                f"THROUGHPUT: {tok_per_sec:,.0f} tok/s"
            )

            if torch.cuda.is_available():
                print(
                    f"GPU MEMORY: "
                    f"{torch.cuda.max_memory_allocated()/1e9:.2f} GB"
                )

            # Save experiment log
            record = {
                "step": step + 1,
                "tokens": int(tokens_seen),
                "train_loss": float(
                    np.mean(history[-50:])
                ),
                "val_mean": val_mean,
                "lr": float(lr),
                "grad_norm": float(grad_norm),
                "tok_per_sec": float(tok_per_sec),
            }

            record.update(val)

            with open(
                f"{RUN}/log.jsonl",
                "a"
            ) as f:
                f.write(
                    json.dumps(record)
                    + "\n"
                )

            # Best checkpoint
            if val_mean < best:

                best = val_mean

                save_ckpt(
                    f"{RUN}/best.pt",
                    step + 1,
                    best,
                    full=False
                )

                print(
                    f"BEST CHECKPOINT SAVED"
                    f" | val={best:.4f}"
                )

            # Latest checkpoint
            save_ckpt(
                f"{RUN}/latest.pt",
                step + 1,
                best
            )

            print(
                f"LATEST CHECKPOINT SAVED"
                f" | step={step+1:,}"
            )

            last_log_time = now
            last_log_step = step + 1

        # ----------------------------------------------------
        # TIME GUARD
        # ----------------------------------------------------

        if timeout:

            print()
            print("=" * 70)
            print("8.25-HOUR TIME GUARD REACHED")
            print("=" * 70)
            print(
                "Checkpoint saved. Resume with the same RUN path."
            )

            return "PAUSED"

    print()
    print("=" * 70)
    print("TRAINING COMPLETE")
    print("=" * 70)

    return "DONE"

# ============================================================
# 13. START REAL RUN
# ============================================================

print()
print("STARTING ARCHITECTURE 3 REAL TRAINING...")
print()

status = run_training()

print()
print("=" * 70)
print("FINAL STATUS:", status)
print("=" * 70)
print("RUN DIRECTORY:", RUN)

for f in [
    "best.pt",
    "latest.pt",
    "log.jsonl"
]:
    path = os.path.join(RUN, f)
    print(
        f"{f:12s}:",
        "OK" if os.path.exists(path) else "MISSING"
    )
```
