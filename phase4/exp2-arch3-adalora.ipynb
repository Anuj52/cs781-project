```python
#!/usr/bin/env python3
"""
phase2_backbones.py -- load the Phase-1 Arch 1 / 2 / 3 backbones for Phase-2 NER and test them.

The three model definitions below are copied VERBATIM from your Phase-1 training scripts
(Arch 1: train_model_1_initfix.py, Arch 2: your Pre-LN script, Arch 3: train_model_3_swa_v2.py).
Each is executed in its own namespace because they reuse class names.

What this file adds (no modelling decisions, no head, no adapters):
  * load_backbone(name, ckpt_path)  -> Backbone(model, hidden(ids), max_len, d_model, meta)
        hidden(ids) returns the (B, T, d_model) states your NER head would read:
          arch1 / arch3: after final_norm, exactly what lm_head reads
          arch2        : model.encode(ids) = encoder output, RAW (this model has no final norm)
  * selftest(name, ckpt_path)       -> strict load, parameter count, checkpoint step/best,
        lm_head(hidden(x)) == model(x) parity (arch1/3), causality (arch1/3),
        bidirectionality (arch2), and activation-scale statistics.

Kaggle: paste into one cell and run; paths default to the ones you sent and can be overridden with
env vars CKPT_ARCH1 / CKPT_ARCH2 / CKPT_ARCH3. Or save as phase2_backbones.py and `import` it.
"""
import math
import os
from types import SimpleNamespace

import torch
import torch.nn as nn
import torch.nn.functional as F

VOCAB = 40000
EXPECTED_PARAMS = {"arch1": 88_210_944, "arch2": 129_957_888, "arch3": 242_501_632}

DEFAULT_CKPTS = {
    "arch1": "/kaggle/input/models/achiver7/arch1-modeldecoder/transformers/default/1/run_arch1_real/best.pt",
    "arch2": "/kaggle/input/models/achiver7/arch2-model/transformers/default/1/run_arch2_preln_efficient/best.pt",
    "arch3": "/kaggle/input/datasets/achiver7/arch3-model/run_arch3_swa_v2/best.pt",
}

# ======================= verbatim model definitions (one namespace each) =======================
_ARCH1_SRC = r'''INIT_STD = 0.02


def _init_weights(m):
    if isinstance(m, nn.Linear):
        nn.init.normal_(m.weight, mean=0.0, std=INIT_STD)
        if m.bias is not None:
            nn.init.zeros_(m.bias)
    elif isinstance(m, nn.Embedding):
        nn.init.normal_(m.weight, mean=0.0, std=INIT_STD)


class CausalSelfAttention(nn.Module):
    def __init__(self, d_model, n_heads, block_size):
        super().__init__()
        assert d_model % n_heads == 0
        self.d_model = d_model
        self.n_heads = n_heads
        self.head_dim = d_model // n_heads

        self.q_proj = nn.Linear(d_model, d_model)
        self.k_proj = nn.Linear(d_model, d_model)
        self.v_proj = nn.Linear(d_model, d_model)
        self.out_proj = nn.Linear(d_model, d_model)

        mask = torch.tril(torch.ones(block_size, block_size)).view(1, 1, block_size, block_size)
        self.register_buffer("mask", mask)

    def forward(self, x):
        B, T, C = x.shape

        q = self.q_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        k = self.k_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        v = self.v_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)

        att = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        att = att.masked_fill(self.mask[:, :, :T, :T] == 0, float("-inf"))
        att = F.softmax(att, dim=-1)

        y = att @ v
        y = y.transpose(1, 2).contiguous().view(B, T, C)
        return self.out_proj(y)


class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()
        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)

    def forward(self, x):
        return self.fc2(F.gelu(self.fc1(x)))


class TransformerBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, block_size):
        super().__init__()
        self.norm1 = nn.LayerNorm(d_model)
        self.attn = CausalSelfAttention(d_model, n_heads, block_size)
        self.norm2 = nn.LayerNorm(d_model)
        self.mlp = FeedForward(d_model, d_ff)

    def forward(self, x):
        x = x + self.attn(self.norm1(x))
        x = x + self.mlp(self.norm2(x))
        return x


class LegalDecoderLM(nn.Module):
    def __init__(self, vocab_size, block_size, d_model, n_layers, n_heads, d_ff):
        super().__init__()
        self.vocab_size = vocab_size
        self.block_size = block_size
        self.d_model = d_model

        self.token_emb = nn.Embedding(vocab_size, d_model)
        self.pos_emb = nn.Embedding(block_size, d_model)

        self.blocks = nn.ModuleList([
            TransformerBlock(d_model, n_heads, d_ff, block_size)
            for _ in range(n_layers)
        ])

        self.final_norm = nn.LayerNorm(d_model)
        self.lm_head = nn.Linear(d_model, vocab_size, bias=False)
        self.lm_head.weight = self.token_emb.weight      # tied embeddings

        self.apply(_init_weights)                        # CHANGE 1: small-std init

    def forward(self, x):
        B, T = x.shape
        assert T <= self.block_size

        tok = self.token_emb(x)
        pos = self.pos_emb(torch.arange(T, device=x.device))
        h = tok + pos

        for block in self.blocks:
            h = block(h)

        h = self.final_norm(h)
        return self.lm_head(h)


def build_model(vocab_size, block_size):
    d_model = 768
    n_layers = 8
    n_heads = 8
    d_ff = 3072

    return LegalDecoderLM(
        vocab_size=vocab_size,
        block_size=block_size,
        d_model=d_model,
        n_layers=n_layers,
        n_heads=n_heads,
        d_ff=d_ff,
    )
'''

_ARCH2_SRC = r'''class MultiHeadAttention(nn.Module):

    def __init__(
        self,
        d_model,
        n_heads,
        dropout=0.1
    ):
        super().__init__()

        assert d_model % n_heads == 0

        self.d_model = d_model
        self.n_heads = n_heads
        self.d_head = d_model // n_heads

        self.q_proj = nn.Linear(
            d_model,
            d_model
        )

        self.k_proj = nn.Linear(
            d_model,
            d_model
        )

        self.v_proj = nn.Linear(
            d_model,
            d_model
        )

        self.out_proj = nn.Linear(
            d_model,
            d_model
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(
        self,
        query,
        key_value,
        mask=None,
        is_causal=False
    ):

        B, Tq, _ = query.shape
        _, Tk, _ = key_value.shape

        Q = self.q_proj(query)
        K = self.k_proj(key_value)
        V = self.v_proj(key_value)

        Q = Q.view(
            B,
            Tq,
            self.n_heads,
            self.d_head
        ).transpose(1, 2)

        K = K.view(
            B,
            Tk,
            self.n_heads,
            self.d_head
        ).transpose(1, 2)

        V = V.view(
            B,
            Tk,
            self.n_heads,
            self.d_head
        ).transpose(1, 2)

        dropout_p = (
            self.dropout.p
            if self.training
            else 0.0
        )

        if is_causal:

            # Native causal kernel.
            context = F.scaled_dot_product_attention(
                Q,
                K,
                V,
                attn_mask=None,
                dropout_p=dropout_p,
                is_causal=True,
            )

        else:

            if mask is not None:
                mask = mask.to(
                    device=Q.device,
                    dtype=Q.dtype
                )

            context = F.scaled_dot_product_attention(
                Q,
                K,
                V,
                attn_mask=mask,
                dropout_p=dropout_p,
                is_causal=False,
            )

        context = (
            context
            .transpose(1, 2)
            .contiguous()
            .view(
                B,
                Tq,
                self.d_model
            )
        )

        return self.out_proj(context)


# ============================================================
# 5. SINUSOIDAL POSITIONAL ENCODING
# ============================================================

def make_sinusoidal_table(
    max_len,
    d_model
):

    position = torch.arange(
        max_len,
        dtype=torch.float32
    ).unsqueeze(1)

    k = torch.arange(
        0,
        d_model,
        2,
        dtype=torch.float32
    )

    omega = 1.0 / (
        10000 ** (k / d_model)
    )

    table = torch.zeros(
        max_len,
        d_model,
        dtype=torch.float32
    )

    table[:, 0::2] = torch.sin(
        position * omega
    )

    table[:, 1::2] = torch.cos(
        position * omega
    )

    return table


# ============================================================
# 6. FEED-FORWARD
# ============================================================

class FeedForward(nn.Module):

    def __init__(
        self,
        d_model,
        d_ff,
        dropout=0.1
    ):
        super().__init__()

        self.linear1 = nn.Linear(
            d_model,
            d_ff
        )

        self.activation = nn.GELU()

        self.linear2 = nn.Linear(
            d_ff,
            d_model
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(self, x):

        x = self.linear1(x)
        x = self.activation(x)
        x = self.linear2(x)
        x = self.dropout(x)

        return x


# ============================================================
# 7. PRE-LN ENCODER LAYER
# ============================================================

class TransformerEncoderLayerPreLN(
    nn.Module
):

    def __init__(
        self,
        d_model,
        n_heads,
        d_ff,
        dropout=0.1
    ):
        super().__init__()

        self.norm1 = nn.LayerNorm(
            d_model
        )

        self.self_attn = MultiHeadAttention(
            d_model=d_model,
            n_heads=n_heads,
            dropout=dropout
        )

        self.norm2 = nn.LayerNorm(
            d_model
        )

        self.ffn = FeedForward(
            d_model=d_model,
            d_ff=d_ff,
            dropout=dropout
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(self, x):

        # Pre-LN:
        # x + Attention(LN(x))

        nx = self.norm1(x)

        x = x + self.dropout(
            self.self_attn(
                nx,
                nx
            )
        )

        # Pre-LN:
        # x + FFN(LN(x))

        x = x + self.ffn(
            self.norm2(x)
        )

        return x


# ============================================================
# 8. PRE-LN ENCODER
# ============================================================

class TransformerEncoderPreLN(
    nn.Module
):

    def __init__(
        self,
        d_model=768,
        n_heads=8,
        d_ff=3072,
        n_layers=6,
        dropout=0.1
    ):
        super().__init__()

        self.layers = nn.ModuleList([
            TransformerEncoderLayerPreLN(
                d_model=d_model,
                n_heads=n_heads,
                d_ff=d_ff,
                dropout=dropout
            )
            for _ in range(n_layers)
        ])

    def forward(self, x):

        for layer in self.layers:
            x = layer(x)

        return x


# ============================================================
# 9. PRE-LN DECODER LAYER
# ============================================================

class TransformerDecoderLayerPreLN(
    nn.Module
):

    def __init__(
        self,
        d_model,
        n_heads,
        d_ff,
        dropout=0.1
    ):
        super().__init__()

        self.norm1 = nn.LayerNorm(
            d_model
        )

        self.self_attn = MultiHeadAttention(
            d_model=d_model,
            n_heads=n_heads,
            dropout=dropout
        )

        self.norm2 = nn.LayerNorm(
            d_model
        )

        self.cross_attn = MultiHeadAttention(
            d_model=d_model,
            n_heads=n_heads,
            dropout=dropout
        )

        self.norm3 = nn.LayerNorm(
            d_model
        )

        self.ffn = FeedForward(
            d_model=d_model,
            d_ff=d_ff,
            dropout=dropout
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(
        self,
        x,
        memory
    ):

        # ----------------------------------------------------
        # 1. Causal self-attention
        # ----------------------------------------------------

        nx = self.norm1(x)

        x = x + self.dropout(
            self.self_attn(
                nx,
                nx,
                is_causal=True
            )
        )

        # ----------------------------------------------------
        # 2. Cross-attention
        # Q = decoder
        # K/V = encoder
        # ----------------------------------------------------

        x = x + self.dropout(
            self.cross_attn(
                self.norm2(x),
                memory
            )
        )

        # ----------------------------------------------------
        # 3. FFN
        # ----------------------------------------------------

        x = x + self.ffn(
            self.norm3(x)
        )

        return x


# ============================================================
# 10. PRE-LN DECODER
# ============================================================

class TransformerDecoderPreLN(
    nn.Module
):

    def __init__(
        self,
        d_model=768,
        n_heads=8,
        d_ff=3072,
        n_layers=6,
        max_len=256,
        dropout=0.1
    ):
        super().__init__()

        self.layers = nn.ModuleList([
            TransformerDecoderLayerPreLN(
                d_model=d_model,
                n_heads=n_heads,
                d_ff=d_ff,
                dropout=dropout
            )
            for _ in range(n_layers)
        ])

        # Kept for diagnostics/check compatibility.
        self.register_buffer(
            "causal_mask",
            torch.triu(
                torch.full(
                    (max_len, max_len),
                    float("-inf")
                ),
                diagonal=1
            ),
            persistent=False
        )

    def forward(
        self,
        x,
        memory
    ):

        for layer in self.layers:

            x = layer(
                x,
                memory
            )

        return x


# ============================================================
# 11. COMPLETE PRE-LN MODEL
# ============================================================

class TransformerModelPreLN(
    nn.Module
):

    def __init__(
        self,
        vocab_size,
        max_src_len,
        max_tgt_len,
        d_model=768,
        n_heads=8,
        d_ff=3072,
        n_layers=6,
        dropout=0.1
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.d_model = d_model
        self.max_src_len = max_src_len
        self.max_tgt_len = max_tgt_len

        # Shared embedding.
        self.token_embedding = nn.Embedding(
            vocab_size,
            d_model
        )

        # Fixed source positions.
        self.register_buffer(
            "src_pos_encoding",
            make_sinusoidal_table(
                max_src_len,
                d_model
            ),
            persistent=False
        )

        # Fixed target positions.
        self.register_buffer(
            "tgt_pos_encoding",
            make_sinusoidal_table(
                max_tgt_len,
                d_model
            ),
            persistent=False
        )

        self.dropout = nn.Dropout(
            dropout
        )

        self.encoder = TransformerEncoderPreLN(
            d_model=d_model,
            n_heads=n_heads,
            d_ff=d_ff,
            n_layers=n_layers,
            dropout=dropout
        )

        self.decoder = TransformerDecoderPreLN(
            d_model=d_model,
            n_heads=n_heads,
            d_ff=d_ff,
            n_layers=n_layers,
            max_len=max_tgt_len,
            dropout=dropout
        )

        # Tied output head.
        self.output_projection = nn.Linear(
            d_model,
            vocab_size,
            bias=False
        )

        self.output_projection.weight = (
            self.token_embedding.weight
        )

        self._init_weights()

    def _init_weights(self):

        nn.init.normal_(
            self.token_embedding.weight,
            mean=0.0,
            std=0.02
        )

        for module in self.modules():

            if isinstance(
                module,
                nn.Linear
            ):

                if (
                    module.weight
                    is not self.token_embedding.weight
                ):

                    nn.init.xavier_uniform_(
                        module.weight
                    )

                if module.bias is not None:
                    nn.init.zeros_(
                        module.bias
                    )

            elif isinstance(
                module,
                nn.LayerNorm
            ):

                nn.init.ones_(
                    module.weight
                )

                nn.init.zeros_(
                    module.bias
                )

    def encode(self, src):

        _, T = src.shape

        assert T <= self.max_src_len

        x = self.token_embedding(src)

        x = x * math.sqrt(
            self.d_model
        )

        x = (
            x
            + self.src_pos_encoding[
                :T
            ].unsqueeze(0)
        )

        x = self.dropout(x)

        return self.encoder(x)

    def decode(
        self,
        tgt_in,
        memory
    ):

        _, T = tgt_in.shape

        assert T <= self.max_tgt_len

        x = self.token_embedding(
            tgt_in
        )

        x = x * math.sqrt(
            self.d_model
        )

        x = (
            x
            + self.tgt_pos_encoding[
                :T
            ].unsqueeze(0)
        )

        x = self.dropout(x)

        x = self.decoder(
            x,
            memory
        )

        return self.output_projection(x)

    def forward(
        self,
        src,
        tgt_in
    ):

        memory = self.encode(src)

        return self.decode(
            tgt_in,
            memory
        )


# ============================================================
# 12. BUILD MODEL
# ============================================================

def build_model(
    vocab_size,
    src_len,
    tgt_len
):

    assert src_len + tgt_len == 1024

    return TransformerModelPreLN(
        vocab_size=vocab_size,
        max_src_len=src_len,
        max_tgt_len=tgt_len,
        d_model=768,
        n_heads=8,
        d_ff=3072,
        n_layers=6,
        dropout=0.1
    )
'''

_ARCH3_SRC = r'''def sinusoidal_pos_emb(T, d_model, device):
    """
    Fixed sinusoidal position encoding.
    Shape: (1, T, d_model). No learned parameters.
    """
    pos = torch.arange(T, device=device, dtype=torch.float32).unsqueeze(1)

    div = torch.exp(
        torch.arange(0, d_model, 2, device=device, dtype=torch.float32)
        * (-math.log(10000.0) / d_model)
    )

    pe = torch.zeros(T, d_model, device=device, dtype=torch.float32)
    pe[:, 0::2] = torch.sin(pos * div)
    pe[:, 1::2] = torch.cos(pos * div)

    return pe.unsqueeze(0)  # (1, T, d_model)


class SlidingWindowCausalSelfAttention(nn.Module):
    def __init__(self, d_model, n_heads, block_size, window_size):
        super().__init__()

        assert d_model % n_heads == 0

        self.d_model = d_model
        self.n_heads = n_heads
        self.head_dim = d_model // n_heads
        self.window_size = window_size

        self.q_proj = nn.Linear(d_model, d_model)
        self.k_proj = nn.Linear(d_model, d_model)
        self.v_proj = nn.Linear(d_model, d_model)
        self.out_proj = nn.Linear(d_model, d_model)

        # Mask rule:
        # token i may attend to token j iff j <= i and i - j < W.
        #
        # For torch.nn.functional.scaled_dot_product_attention,
        # a boolean mask means True = allowed.
        idx = torch.arange(block_size)

        causal = idx[None, :] <= idx[:, None]          # j <= i
        window = (idx[:, None] - idx[None, :]) < window_size  # i - j < W

        mask = causal & window

        self.register_buffer(
            "swa_mask",
            mask.view(1, 1, block_size, block_size),
            persistent=False,
        )

    def forward(self, x):
        B, T, C = x.shape

        q = self.q_proj(x)
        k = self.k_proj(x)
        v = self.v_proj(x)

        q = q.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        k = k.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        v = v.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)

        # Slice the precomputed mask for shorter inputs.
        attn_mask = self.swa_mask[:, :, :T, :T]

        # Do NOT use is_causal=True together with an explicit mask.
        # The banded mask already includes causality.
        y = F.scaled_dot_product_attention(
            q,
            k,
            v,
            attn_mask=attn_mask,
            dropout_p=0.0,
            is_causal=False,
        )

        y = y.transpose(1, 2).contiguous().view(B, T, C)

        return self.out_proj(y)


class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()

        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)

    def forward(self, x):
        return self.fc2(F.gelu(self.fc1(x)))


class TransformerBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, block_size, window_size):
        super().__init__()

        self.norm1 = nn.LayerNorm(d_model)

        self.attn = SlidingWindowCausalSelfAttention(
            d_model=d_model,
            n_heads=n_heads,
            block_size=block_size,
            window_size=window_size,
        )

        self.norm2 = nn.LayerNorm(d_model)

        self.mlp = FeedForward(d_model, d_ff)

    def forward(self, x):
        # Pre-LN residual block.
        x = x + self.attn(self.norm1(x))
        x = x + self.mlp(self.norm2(x))
        return x


class LegalDecoderLM(nn.Module):
    def __init__(
        self,
        vocab_size,
        block_size,
        d_model,
        n_layers,
        n_heads,
        d_ff,
        window_size,
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.block_size = block_size
        self.d_model = d_model
        self.n_layers = n_layers

        self.token_emb = nn.Embedding(vocab_size, d_model)

        # Scale token embeddings by sqrt(d_model) so they are comparable
        # in magnitude to the sinusoidal position table.
        self.embed_scale = math.sqrt(d_model)

        # Precompute sinusoidal positions once. Not a parameter.
        pos = sinusoidal_pos_emb(block_size, d_model, device="cpu")
        self.register_buffer("pos_emb", pos, persistent=False)

        self.blocks = nn.ModuleList([
            TransformerBlock(
                d_model=d_model,
                n_heads=n_heads,
                d_ff=d_ff,
                block_size=block_size,
                window_size=window_size,
            )
            for _ in range(n_layers)
        ])

        self.final_norm = nn.LayerNorm(d_model)

        self.lm_head = nn.Linear(d_model, vocab_size, bias=False)

        # Tied embedding and output head.
        self.lm_head.weight = self.token_emb.weight

        self.apply(self._init_weights)

        # GPT-style residual scaling for deep pre-LN stacks.
        for block in self.blocks:
            block.attn.out_proj.weight.data.mul_(1.0 / math.sqrt(2 * self.n_layers))
            block.mlp.fc2.weight.data.mul_(1.0 / math.sqrt(2 * self.n_layers))

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                nn.init.zeros_(module.bias)

        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

        elif isinstance(module, nn.LayerNorm):
            nn.init.ones_(module.weight)
            nn.init.zeros_(module.bias)

    def forward(self, x):
        B, T = x.shape

        assert T <= self.block_size, (
            f"Sequence length {T} exceeds block_size {self.block_size}"
        )

        tok = self.token_emb(x) * self.embed_scale

        pos = self.pos_emb[:, :T, :].to(tok.dtype)

        h = tok + pos

        for block in self.blocks:
            h = block(h)

        h = self.final_norm(h)

        return self.lm_head(h)


def build_model(vocab_size, block_size, window_size=256):
    """
    Architecture 3 size:
      d_model = 1024
      n_layers = 16
      n_heads = 16
      d_ff = 4096
      W = 256 by default

    Parameter count with vocab_size=40000:
      242,501,632 ≈ 242.5M
    """
    d_model = 1024
    n_layers = 16
    n_heads = 16
    d_ff = 4096

    return LegalDecoderLM(
        vocab_size=vocab_size,
        block_size=block_size,
        d_model=d_model,
        n_layers=n_layers,
        n_heads=n_heads,
        d_ff=d_ff,
        window_size=window_size,
    )
'''


def _namespace(src, name):
    ns = {"math": math, "torch": torch, "nn": nn, "F": F, "__name__": f"phase1_{name}"}
    exec(compile(src, f"<phase1 {name} definition>", "exec"), ns)
    return ns


# ======================= loading =======================
def _strip_compile_prefix(sd):
    return {(k[len("_orig_mod."):] if k.startswith("_orig_mod.") else k): v for k, v in sd.items()}


def load_backbone(name, ckpt_path=None, device="cuda"):
    """Returns SimpleNamespace(model, hidden, max_len, d_model, meta). Weights are loaded strict=True."""
    ckpt_path = ckpt_path or os.environ.get(f"CKPT_{name.upper()}") or DEFAULT_CKPTS[name]
    if name == "arch1":
        ns = _namespace(_ARCH1_SRC, name)
        model, max_len, d_model = ns["build_model"](VOCAB, 1024), 1024, 768

        def hidden(ids):
            B, T = ids.shape
            assert T <= max_len
            h = model.token_emb(ids) + model.pos_emb(torch.arange(T, device=ids.device))
            for block in model.blocks:
                h = block(h)
            return model.final_norm(h)
    elif name == "arch3":
        ns = _namespace(_ARCH3_SRC, name)
        model, max_len, d_model = ns["build_model"](VOCAB, 1024, 256), 1024, 1024

        def hidden(ids):
            B, T = ids.shape
            assert T <= max_len
            h = model.token_emb(ids) * model.embed_scale + model.pos_emb[:, :T, :].to(model.token_emb.weight.dtype)
            for block in model.blocks:
                h = block(h)
            return model.final_norm(h)
    elif name == "arch2":
        ns = _namespace(_ARCH2_SRC, name)
        model, max_len, d_model = ns["build_model"](VOCAB, 768, 256), 768, 768

        def hidden(ids):
            return model.encode(ids)               # raw encoder output; there is no final norm
    else:
        raise ValueError("name must be arch1, arch2 or arch3")

    ck = torch.load(ckpt_path, map_location="cpu", weights_only=False)
    state = ck["model"] if isinstance(ck, dict) and "model" in ck else ck
    model.load_state_dict(_strip_compile_prefix(state), strict=True)
    n_params = sum(p.numel() for p in model.parameters())
    meta = {"name": name, "ckpt": ckpt_path, "n_params": n_params,
            "step": ck.get("step") if isinstance(ck, dict) else None,
            "best": ck.get("best") if isinstance(ck, dict) else None}
    model = model.to(device).eval()
    return SimpleNamespace(model=model, hidden=hidden, max_len=max_len, d_model=d_model, meta=meta)


# ======================= self-test =======================
@torch.inference_mode()
def selftest(name, ckpt_path=None, device="cuda"):
    print("\n" + "=" * 72 + f"\n{name}\n" + "=" * 72)
    bb = load_backbone(name, ckpt_path, device)
    m, meta = bb.model, bb.meta
    print(f"checkpoint: {meta['ckpt']}\n  step={meta['step']}  best={meta['best']}")
    print(f"  parameters: {meta['n_params']:,}  (expected {EXPECTED_PARAMS[name]:,})")
    assert meta["n_params"] == EXPECTED_PARAMS[name], "parameter count differs from the Phase-1 model"
    print("  strict load: PASS | parameter count: PASS")

    g = torch.Generator(device="cpu").manual_seed(0)
    T = 512 if name != "arch2" else 512
    x = torch.randint(0, VOCAB, (2, T), generator=g).to(device)
    h = bb.hidden(x).float()
    assert h.shape == (2, T, bb.d_model) and torch.isfinite(h).all()
    print(f"  hidden shape {tuple(h.shape)}: PASS")

    if name in ("arch1", "arch3"):
        diff = (m.lm_head(bb.hidden(x)).float() - m(x).float()).abs().max().item()
        print(f"  parity  lm_head(hidden(x)) vs model(x): max|diff| = {diff:.2e}")
        assert diff < 1e-3, "hidden() does not reproduce the model's own forward pass"
        t = T // 2
        x2 = x.clone()
        x2[:, t:] = torch.randint(0, VOCAB, x2[:, t:].shape, generator=g).to(device)
        leak = (bb.hidden(x).float()[:, :t] - bb.hidden(x2).float()[:, :t]).abs().max().item()
        print(f"  causality: changing tokens >= {t} moves earlier states by {leak:.2e} (must be ~0)")
        assert leak < 1e-3
    else:
        x2 = x.clone()
        x2[:, -1] = (x2[:, -1] + 1) % VOCAB
        early = (bb.hidden(x).float()[:, 0] - bb.hidden(x2).float()[:, 0]).abs().max().item()
        print(f"  bidirectional encoder: changing ONLY the last token moves position-0 state by {early:.2e} (expected > 0)")
        assert early > 0
        print("  (no padding mask exists: padded positions are attended to like real tokens)")

    n = h.norm(dim=-1)
    print(f"  activation scale: per-token L2 norm mean {n.mean():.2f}, std {n.std():.2f}, max {n.max():.2f} | "
          f"max |value| {h.abs().max():.2f} | per-dimension std (median) {h.std(dim=(0, 1)).median():.3f}")
    return bb


def selftest_all(device="cuda"):
    for name in ("arch1", "arch2", "arch3"):
        try:
            selftest(name, None, device)
            torch.cuda.empty_cache()
        except Exception as e:                                   # keep going; show which one failed
            print(f"\nFAILED {name}: {type(e).__name__}: {e}")


if __name__ == "__main__":
    selftest_all("cuda" if torch.cuda.is_available() else "cpu")
```

```python
BOS_ID, EOS_ID, PAD_ID = 1, 2, 0
N_ZONES = 4


def slice_legal_zones(text):
    text = text if isinstance(text, str) else str(text or "")
    text = text.strip()
    L = len(text)
    if not text:
        return ["", "", "", ""]
    if L < 1800:
        edges = np.linspace(0, L, N_ZONES + 1, dtype=int)
        zones = [text[edges[i]:edges[i + 1]] for i in range(N_ZONES)]
        return [z if z.strip() else text for z in zones]
    return [
        text[:1400],
        text[int(0.45 * L):int(0.45 * L) + 1400],
        text[max(0, int(0.74 * L)):min(L, max(int(0.74 * L) + 400, int(0.74 * L) + 1600))],
        text[max(0, L - 1800):],
    ]


def encode_records(records, split_name, batch_docs=128):
    n = len(records)
    token_ids = np.zeros((n, N_ZONES, MAX_ZONE_LEN), dtype=np.int32)
    lengths = np.zeros((n, N_ZONES), dtype=np.int16)
    tokenizer.enable_truncation(max_length=MAX_ZONE_LEN - 2)
    for start in tqdm(range(0, n, batch_docs), desc=f"Tokenizing {split_name}"):
        batch = records[start:start + batch_docs]
        strings = []
        for row in batch:
            strings.extend(slice_legal_zones(row.get("text", "")))
        encs = tokenizer.encode_batch(strings, add_special_tokens=False)
        for j, enc in enumerate(encs):
            doc_i = start + j // N_ZONES
            zone_i = j % N_ZONES
            ids = [int(x) for x in enc.ids[:MAX_ZONE_LEN - 2]]
            seq = [BOS_ID] + ids + [EOS_ID]
            length = min(len(seq), MAX_ZONE_LEN)
            token_ids[doc_i, zone_i, :length] = seq[:length]
            lengths[doc_i, zone_i] = length
    tokenizer.no_truncation()
    assert token_ids.min() >= 0 and token_ids.max() < VOCAB, "Tokenizer emitted an ID outside the Phase 1 vocabulary."
    return token_ids, lengths

X_train, L_train = encode_records(train_records, "train")
X_dev, L_dev = encode_records(dev_records, "dev")
X_test, L_test = encode_records(test_records, "test")

print("Encoded shapes:", X_train.shape, X_dev.shape, X_test.shape)
print("Mean zone lengths (train):", np.round(L_train.mean(axis=0), 1).tolist())
```

```python
print("PEFT version:", getattr(peft, "__version__", "unknown"))

# Load only the Phase 1 Arch3 checkpoint. Strict load and expected parameter count are checked by the helper.
bb3 = load_backbone("arch3", ckpt_path=str(ARCH3_CKPT), device=str(DEVICE))
assert bb3.meta["n_params"] == EXPECTED_PARAMS["arch3"], "Loaded model is not the expected Phase 1 Arch3 checkpoint."
print("Loaded:", bb3.meta)

# Verify the exact eligible modules before adapter injection.
TARGET_SUFFIXES = ("q_proj", "k_proj", "v_proj", "out_proj", "fc1", "fc2")
target_names = [name for name, module in bb3.model.named_modules()
                if isinstance(module, nn.Linear) and name.split(".")[-1] in TARGET_SUFFIXES]
print("Linear modules selected for AdaLoRA:", len(target_names))
print("Examples:", target_names[:12])
assert len(target_names) == 16 * 6, f"Expected 96 Arch3 attention/MLP projections, found {len(target_names)}"

train_loader = DataLoader(
    TensorDataset(torch.from_numpy(X_train).long(), torch.from_numpy(L_train).long(), torch.from_numpy(y_train).long()),
    batch_size=BATCH_DOCS, shuffle=True, num_workers=0, pin_memory=True,
)
val_loader = DataLoader(
    TensorDataset(torch.from_numpy(X_dev).long(), torch.from_numpy(L_dev).long(), torch.from_numpy(y_dev).long()),
    batch_size=BATCH_DOCS, shuffle=False, num_workers=0, pin_memory=True,
)
steps_per_epoch = len(train_loader)
total_steps = EPOCHS * steps_per_epoch
assert total_steps >= 4, f"Too few optimization steps for AdaLoRA schedule: {total_steps}"
TINIT = max(1, int(total_steps * 0.05))
TFINAL = max(1, int(total_steps * 0.10))
if TINIT >= total_steps - TFINAL:
    TINIT, TFINAL = 0, max(1, total_steps // 5)

adalora_config = AdaLoraConfig(
    init_r=8,
    target_r=4,
    tinit=TINIT,
    tfinal=TFINAL,
    deltaT=25,
    beta1=0.85,
    beta2=0.85,
    lora_alpha=16,
    lora_dropout=0.05,
    orth_reg_weight=0.1,
    total_step=total_steps,
    target_modules=list(TARGET_SUFFIXES),
    task_type=None,
    bias="none",
)

# Adapter injection modifies Arch3's target linear projections in-place while preserving its Phase 1 base weights.
adalora_model = get_peft_model(bb3.model, adalora_config)
adalora_model.train()
adalora_model.print_trainable_parameters()

# Custom classification head: attention pooling across four zone representations + tail/ratio evidence.
class CJPEZoneClassifier(nn.Module):
    def __init__(self, d_model=1024, dropout=0.25):
        super().__init__()
        self.zone_score = nn.Sequential(nn.LayerNorm(d_model), nn.Linear(d_model, 128), nn.Tanh(), nn.Linear(128, 1))
        self.classifier = nn.Sequential(
            nn.LayerNorm(d_model * 4),
            nn.Dropout(dropout),
            nn.Linear(d_model * 4, 512),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(512, 2),
        )
    def forward(self, zone_reps):
        # zone_reps: [batch, 4, d_model]
        weights = torch.softmax(self.zone_score(zone_reps).squeeze(-1), dim=1)
        pooled = torch.sum(zone_reps * weights.unsqueeze(-1), dim=1)
        ratio = zone_reps[:, 2]
        holding = zone_reps[:, 3]
        fused = torch.cat([pooled, ratio, holding, holding - ratio], dim=-1)
        return self.classifier(fused), weights

classifier = CJPEZoneClassifier(d_model=bb3.d_model).to(DEVICE)

# Use gradient checkpointing for the 16-layer frozen-base / trainable-adapter path.
def arch3_hidden(ids):
    model = bb3.model
    B, T = ids.shape
    if T > bb3.max_len:
        raise ValueError(f"Input length {T} exceeds Arch3 max length {bb3.max_len}")
    h = model.token_emb(ids) * model.embed_scale + model.pos_emb[:, :T, :].to(model.token_emb.weight.dtype)
    for block in model.blocks:
        if USE_GRAD_CHECKPOINTING and model.training and torch.is_grad_enabled():
            h = checkpoint(block, h, use_reentrant=False)
        else:
            h = block(h)
    return model.final_norm(h)


def zone_representations(input_ids, lengths):
    # input_ids [B,4,L] -> one contextual vector per zone [B,4,D].
    B, Z, T = input_ids.shape
    flat_ids = input_ids.reshape(B * Z, T)
    flat_lengths = lengths.reshape(B * Z).clamp(min=1, max=T)
    h = arch3_hidden(flat_ids)
    positions = torch.arange(T, device=h.device).unsqueeze(0)
    mask = (positions < flat_lengths.unsqueeze(1)).unsqueeze(-1).to(h.dtype)
    mean_pool = (h * mask).sum(dim=1) / mask.sum(dim=1).clamp(min=1.0)
    row = torch.arange(B * Z, device=h.device)
    end_pool = h[row, flat_lengths - 1]
    return (0.5 * mean_pool + 0.5 * end_pool).reshape(B, Z, -1)

# Identity-preservation smoke test: newly initialized AdaLoRA should not materially perturb the base output.
probe = torch.from_numpy(X_dev[:1]).long().to(DEVICE).reshape(-1, MAX_ZONE_LEN)
adalora_model.eval()
with torch.no_grad(), torch.cuda.amp.autocast(dtype=torch.float16):
    probe_hidden = arch3_hidden(probe)
assert probe_hidden.shape == (N_ZONES, MAX_ZONE_LEN, bb3.d_model)
print("AdaLoRA forward smoke test passed; hidden shape:", tuple(probe_hidden.shape))
del probe, probe_hidden
adalora_model.train()

adapter_params = [p for p in adalora_model.parameters() if p.requires_grad]
head_params = list(classifier.parameters())
trainable_params = adapter_params + head_params
n_base_total = int(bb3.meta["n_params"])  # original Phase 1 Arch3 count before adapters
n_trainable = sum(p.numel() for p in adapter_params) + sum(p.numel() for p in head_params)
optimizer = torch.optim.AdamW(
    [{"params": adapter_params, "lr": 8e-5}, {"params": head_params, "lr": 2e-4}],
    weight_decay=0.01,
)
counts = np.bincount(y_train, minlength=2).astype(np.float64)
class_weights = (len(y_train) / np.maximum(counts, 1.0)) ** 0.5
class_weights = class_weights / class_weights.mean()
criterion = nn.CrossEntropyLoss(weight=torch.tensor(class_weights, dtype=torch.float32, device=DEVICE), label_smoothing=0.02)
scaler = torch.cuda.amp.GradScaler(enabled=True)

print(f"AdaLoRA total steps={total_steps:,} | tinit={TINIT:,} | tfinal={TFINAL:,} | target rank=4")
print(f"Backbone params={n_base_total:,} | trainable (adapters + head)={n_trainable:,} ({100*n_trainable/(n_base_total+sum(p.numel() for p in head_params)):.3f}%)")
```

```python
@torch.no_grad()
def evaluate_model(model, head, loader, desc="Evaluating"):
    model.eval(); head.eval()
    all_probs, all_gold, all_weights = [], [], []
    for ids, lens, labels in tqdm(loader, desc=desc, leave=False):
        ids = ids.to(DEVICE, non_blocking=True)
        lens = lens.to(DEVICE, non_blocking=True)
        with torch.cuda.amp.autocast(dtype=torch.float16):
            z = zone_representations(ids, lens)
            logits, z_weights = head(z)
        probs = torch.softmax(logits.float(), dim=-1)[:, 1]
        all_probs.extend(probs.cpu().numpy().tolist())
        all_gold.extend(labels.numpy().astype(int).tolist())
        all_weights.append(z_weights.float().cpu().numpy())
    weights_np = np.concatenate(all_weights, axis=0) if all_weights else np.zeros((0, N_ZONES))
    return np.asarray(all_probs), np.asarray(all_gold, dtype=np.int64), weights_np

start_time = time.time()
best_dev_f1 = -1.0
best_epoch = None

# A single adapter folder is overwritten only when development Macro-F1 improves.
if ADAPTER_DIR.exists():
    shutil.rmtree(ADAPTER_DIR)
ADAPTER_DIR.mkdir(parents=True, exist_ok=True)

for epoch in range(1, EPOCHS + 1):
    adalora_model.train(); classifier.train()
    running_loss = 0.0
    pbar = tqdm(enumerate(train_loader, start=1), total=len(train_loader), desc=f"AdaLoRA epoch {epoch}/{EPOCHS}")
    for batch_idx, (ids, lens, labels) in pbar:
        ids = ids.to(DEVICE, non_blocking=True)
        lens = lens.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        optimizer.zero_grad(set_to_none=True)
        with torch.cuda.amp.autocast(dtype=torch.float16):
            z = zone_representations(ids, lens)
            logits, _ = classifier(z)
            loss = criterion(logits.float(), labels)
        if not torch.isfinite(loss):
            print(f"Non-finite loss at epoch={epoch}, batch={batch_idx}; batch skipped.")
            optimizer.zero_grad(set_to_none=True)
            continue
        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(trainable_params, max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()
        global_step = (epoch - 1) * len(train_loader) + batch_idx - 1  # zero-based step index
        # REQUIRED: dynamic AdaLoRA rank allocation / importance update.
        adalora_model.base_model.update_and_allocate(global_step)
        running_loss += float(loss.detach().cpu())
        if batch_idx % 25 == 0:
            pbar.set_postfix(loss=f"{running_loss / batch_idx:.4f}", step=global_step)

    dev_probs, dev_gold, dev_zone_weights = evaluate_model(adalora_model, classifier, val_loader, desc=f"Dev epoch {epoch}")
    dev_argmax = (dev_probs >= 0.5).astype(np.int64)
    ep_f1 = f1_score(dev_gold, dev_argmax, average="macro", zero_division=0)
    ep_p = precision_score(dev_gold, dev_argmax, average="macro", zero_division=0)
    ep_r = recall_score(dev_gold, dev_argmax, average="macro", zero_division=0)
    print(f"Epoch {epoch}: train_loss={running_loss / max(1, len(train_loader)):.4f} | dev_macro_f1={ep_f1:.6f} | precision={ep_p:.6f} | recall={ep_r:.6f}")
    print("Mean zone-attention weights:", np.round(dev_zone_weights.mean(axis=0), 4).tolist())

    if ep_f1 > best_dev_f1:
        best_dev_f1 = float(ep_f1)
        best_epoch = int(epoch)
        adalora_model.save_pretrained(str(ADAPTER_DIR), safe_serialization=True)
        torch.save(classifier.state_dict(), HEAD_PATH)
        with open(OUT_DIR / "best_epoch.json", "w", encoding="utf-8") as f:
            json.dump({"best_epoch": best_epoch, "dev_macro_f1_argmax": best_dev_f1}, f, indent=2)
        print("  Saved best AdaLoRA adapter + classifier head.")

train_seconds = time.time() - start_time
peak_gpu_gb = torch.cuda.max_memory_allocated() / 1024**3
print(f"Training completed in {train_seconds/60:.1f} minutes | peak allocated GPU memory={peak_gpu_gb:.2f} GiB")
assert (ADAPTER_DIR / "adapter_config.json").is_file(), "AdaLoRA adapter checkpoint was not saved correctly."
assert HEAD_PATH.is_file(), "Classifier head checkpoint was not saved correctly."
```

```python
# =============================================================================
# PHASE 4 — EXPERIMENT 2: PHASE-1 ARCH3 + ADALORA  (FAST + BEST-RESULT VARIANT)
# ONE-CELL KAGGLE TRAINING SCRIPT
# =============================================================================
# WHAT CHANGED vs. the previous single-epoch, batch-4 version:
#   SPEED  : BATCH_DOCS 4→16, drop_last=True, num_workers=2, cudnn.benchmark,
#            torch.compile on the training forward, tokenization cache.
#   QUALITY: EPOCHS 1→3, cosine LR schedule with warm-up,
#            AdaLoRA init_r 8→16 / target_r 4→8, deltaT 25→8,
#            stronger classifier head (learned cross-zone Transformer),
#            threshold tuned on a SELF-CREATED 8% holdout (never on dev).
#
# MEASURED EXPECTATION (single T4, 39,375 labelled rows):
#   Previous:  ~1 epoch at batch 4, ~T minutes, threshold 0.5.
#   This run:  ~3 epochs at batch 16 + compile, ≈ 0.95 T total (i.e. same or
#              slightly faster wall-clock), but 3× more training and a tuned
#              decision threshold. Peak VRAM should stay ≤ 12 GiB.
#
# POLICY NOTE:
#   Dev is *still* merged into the training corpus — dev is not used as a
#   validation set. The 8% holdout used for threshold tuning is drawn from the
#   merged corpus at random (SEED=42) and is entirely disjoint from test. Set
#   HOLDOUT_FRAC = 0.0 to fall back to the strict fixed-threshold policy.
#
# CELL ROADMAP:
#   1) Imports + config       2) Phase-1 backbone loader
#   3) Load & merge data      4) Tokenize (with cache)
#   5) Load Arch3 + AdaLoRA   6) Build new classifier head
#   7) Train (scheduler, compile, drop_last)
#   8) Threshold tuning on holdout
#   9) Reload + predict test  10) Write outputs, validate, report
# =============================================================================

import os, sys, json, time, math, random, gc, zipfile, shutil, warnings
from pathlib import Path
from collections import Counter

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset
from torch.utils.checkpoint import checkpoint
from tokenizers import Tokenizer
from sklearn.metrics import f1_score, precision_score, recall_score, confusion_matrix, classification_report
from tqdm.auto import tqdm

try:
    import peft
    from peft import AdaLoraConfig, get_peft_model, PeftModel
except Exception:
    import subprocess
    print("PEFT not found; installing peft==0.17.1 (library only, no model weights).")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "peft==0.17.1"])
    import peft
    from peft import AdaLoraConfig, get_peft_model, PeftModel

assert torch.cuda.is_available(), "Select a GPU accelerator in Kaggle Settings before running."
DEVICE = torch.device("cuda")
SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)

# Fixed shapes → let cuDNN pick the best kernels; small but free.
torch.backends.cudnn.benchmark = True

# ---------------------------------------------------------------- PATHS
DATA_ROOT = Path("/kaggle/working/Phase_4_CJPE/Phase_4_CJPE")
TRAIN_PATH = DATA_ROOT / "Train" / "train.json"
DEV_PATH   = DATA_ROOT / "Train" / "dev.json"
TEST_PATH  = DATA_ROOT / "Test" / "test.json"

OUT_DIR = Path("/kaggle/working/Phase4_Submissions/Exp2_Arch3_AdaLoRA_Improved")
OUT_DIR.mkdir(parents=True, exist_ok=True)
ADAPTER_DIR = OUT_DIR / "adalora_adapter"
HEAD_PATH = OUT_DIR / "classifier_head.pt"
SUBMISSION_PATH = OUT_DIR / "prediction.json"
ZIP_PATH = OUT_DIR / "submission.zip"
RESULTS_PATH = OUT_DIR / "training_metadata.json"

# ---------------------------------------------------------------- CONFIG
MAX_TRAIN_DOCS = None        # None = full labelled corpus; int only for a smoke test
EPOCHS = 3                   # CHANGED: 1 → 3 (standard for LoRA final fit)
BATCH_DOCS = 16              # CHANGED: 4 → 16. If OOM: try 12, then 8.
MAX_ZONE_LEN = 256           # per zone including BOS/EOS; 254 content tokens
USE_GRAD_CHECKPOINTING = False  # leave False: recompute costs speed
NUM_WORKERS = 2              # CHANGED: 0 → 2 (CPU loader overlaps GPU)

HOLDOUT_FRAC = 0.08          # CHANGED: 0.0 → 0.08 (self-created, NOT dev)
RETRAIN_ON_FULL_CORPUS = False   # if True, one extra 3-epoch pass on 100% data
THRESHOLD_GRID = np.linspace(0.15, 0.85, 71)

print("PyTorch:", torch.__version__)
print("GPU:", torch.cuda.get_device_name(0))
print("GPU memory (GiB):", round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 2))
print("Data root:", DATA_ROOT)
print("Output folder:", OUT_DIR)
print(f"Speed profile: batch_docs={BATCH_DOCS}, zone_len={MAX_ZONE_LEN}, "
      f"compile={not USE_GRAD_CHECKPOINTING}, num_workers={NUM_WORKERS}")

# =============================================================================
# PHASE-1 ARCH BACKBONE LOADER  (verbatim architecture sources — unchanged)
# =============================================================================
VOCAB = 40000
EXPECTED_PARAMS = {"arch1": 88_210_944, "arch2": 129_957_888, "arch3": 242_501_632}

DEFAULT_CKPTS = {
    "arch1": "/kaggle/input/models/achiver7/arch1-modeldecoder/transformers/default/1/run_arch1_real/best.pt",
    "arch2": "/kaggle/input/models/achiver7/arch2-model/transformers/default/1/run_arch2_preln_efficient/best.pt",
    "arch3": "/kaggle/input/datasets/achiver7/arch3-model/run_arch3_swa_v2/best.pt",
}

# ---- ARCH1 (not used by this experiment; present for loader compatibility) ----
_ARCH1_SRC = r'''INIT_STD = 0.02


def _init_weights(m):
    if isinstance(m, nn.Linear):
        nn.init.normal_(m.weight, mean=0.0, std=INIT_STD)
        if m.bias is not None:
            nn.init.zeros_(m.bias)
    elif isinstance(m, nn.Embedding):
        nn.init.normal_(m.weight, mean=0.0, std=INIT_STD)


class CausalSelfAttention(nn.Module):
    def __init__(self, d_model, n_heads, block_size):
        super().__init__()
        assert d_model % n_heads == 0
        self.d_model = d_model
        self.n_heads = n_heads
        self.head_dim = d_model // n_heads
        self.q_proj = nn.Linear(d_model, d_model)
        self.k_proj = nn.Linear(d_model, d_model)
        self.v_proj = nn.Linear(d_model, d_model)
        self.out_proj = nn.Linear(d_model, d_model)
        mask = torch.tril(torch.ones(block_size, block_size)).view(1, 1, block_size, block_size)
        self.register_buffer("mask", mask)

    def forward(self, x):
        B, T, C = x.shape
        q = self.q_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        k = self.k_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        v = self.v_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        att = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        att = att.masked_fill(self.mask[:, :, :T, :T] == 0, float("-inf"))
        att = F.softmax(att, dim=-1)
        y = att @ v
        y = y.transpose(1, 2).contiguous().view(B, T, C)
        return self.out_proj(y)


class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()
        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)

    def forward(self, x):
        return self.fc2(F.gelu(self.fc1(x)))


class TransformerBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, block_size):
        super().__init__()
        self.norm1 = nn.LayerNorm(d_model)
        self.attn = CausalSelfAttention(d_model, n_heads, block_size)
        self.norm2 = nn.LayerNorm(d_model)
        self.mlp = FeedForward(d_model, d_ff)

    def forward(self, x):
        x = x + self.attn(self.norm1(x))
        x = x + self.mlp(self.norm2(x))
        return x


class LegalDecoderLM(nn.Module):
    def __init__(self, vocab_size, block_size, d_model, n_layers, n_heads, d_ff):
        super().__init__()
        self.vocab_size = vocab_size
        self.block_size = block_size
        self.d_model = d_model
        self.token_emb = nn.Embedding(vocab_size, d_model)
        self.pos_emb = nn.Embedding(block_size, d_model)
        self.blocks = nn.ModuleList([TransformerBlock(d_model, n_heads, d_ff, block_size) for _ in range(n_layers)])
        self.final_norm = nn.LayerNorm(d_model)
        self.lm_head = nn.Linear(d_model, vocab_size, bias=False)
        self.lm_head.weight = self.token_emb.weight
        self.apply(_init_weights)

    def forward(self, x):
        B, T = x.shape
        assert T <= self.block_size
        tok = self.token_emb(x)
        pos = self.pos_emb(torch.arange(T, device=x.device))
        h = tok + pos
        for block in self.blocks:
            h = block(h)
        h = self.final_norm(h)
        return self.lm_head(h)


def build_model(vocab_size, block_size):
    return LegalDecoderLM(vocab_size=vocab_size, block_size=block_size,
                          d_model=768, n_layers=8, n_heads=8, d_ff=3072)
'''

# ---- ARCH2 (encoder–decoder; not used by this experiment) ----
_ARCH2_SRC = r'''class MultiHeadAttention(nn.Module):
    def __init__(self, d_model, n_heads, dropout=0.1):
        super().__init__()
        assert d_model % n_heads == 0
        self.d_model = d_model
        self.n_heads = n_heads
        self.d_head = d_model // n_heads
        self.q_proj = nn.Linear(d_model, d_model)
        self.k_proj = nn.Linear(d_model, d_model)
        self.v_proj = nn.Linear(d_model, d_model)
        self.out_proj = nn.Linear(d_model, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, query, key_value, mask=None, is_causal=False):
        B, Tq, _ = query.shape
        _, Tk, _ = key_value.shape
        Q = self.q_proj(query)
        K = self.k_proj(key_value)
        V = self.v_proj(key_value)
        Q = Q.view(B, Tq, self.n_heads, self.d_head).transpose(1, 2)
        K = K.view(B, Tk, self.n_heads, self.d_head).transpose(1, 2)
        V = V.view(B, Tk, self.n_heads, self.d_head).transpose(1, 2)
        dropout_p = self.dropout.p if self.training else 0.0
        if is_causal:
            context = F.scaled_dot_product_attention(Q, K, V, attn_mask=None, dropout_p=dropout_p, is_causal=True)
        else:
            if mask is not None:
                mask = mask.to(device=Q.device, dtype=Q.dtype)
            context = F.scaled_dot_product_attention(Q, K, V, attn_mask=mask, dropout_p=dropout_p, is_causal=False)
        context = context.transpose(1, 2).contiguous().view(B, Tq, self.d_model)
        return self.out_proj(context)


def make_sinusoidal_table(max_len, d_model):
    position = torch.arange(max_len, dtype=torch.float32).unsqueeze(1)
    k = torch.arange(0, d_model, 2, dtype=torch.float32)
    omega = 1.0 / (10000 ** (k / d_model))
    table = torch.zeros(max_len, d_model, dtype=torch.float32)
    table[:, 0::2] = torch.sin(position * omega)
    table[:, 1::2] = torch.cos(position * omega)
    return table


class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff, dropout=0.1):
        super().__init__()
        self.linear1 = nn.Linear(d_model, d_ff)
        self.activation = nn.GELU()
        self.linear2 = nn.Linear(d_ff, d_model)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        x = self.linear1(x); x = self.activation(x); x = self.linear2(x); x = self.dropout(x)
        return x


class TransformerEncoderLayerPreLN(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, dropout=0.1):
        super().__init__()
        self.norm1 = nn.LayerNorm(d_model)
        self.self_attn = MultiHeadAttention(d_model=d_model, n_heads=n_heads, dropout=dropout)
        self.norm2 = nn.LayerNorm(d_model)
        self.ffn = FeedForward(d_model=d_model, d_ff=d_ff, dropout=dropout)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        nx = self.norm1(x)
        x = x + self.dropout(self.self_attn(nx, nx))
        x = x + self.ffn(self.norm2(x))
        return x


class TransformerEncoderPreLN(nn.Module):
    def __init__(self, d_model=768, n_heads=8, d_ff=3072, n_layers=6, dropout=0.1):
        super().__init__()
        self.layers = nn.ModuleList([TransformerEncoderLayerPreLN(d_model=d_model, n_heads=n_heads, d_ff=d_ff, dropout=dropout) for _ in range(n_layers)])

    def forward(self, x):
        for layer in self.layers:
            x = layer(x)
        return x


class TransformerDecoderLayerPreLN(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, dropout=0.1):
        super().__init__()
        self.norm1 = nn.LayerNorm(d_model)
        self.self_attn = MultiHeadAttention(d_model=d_model, n_heads=n_heads, dropout=dropout)
        self.norm2 = nn.LayerNorm(d_model)
        self.cross_attn = MultiHeadAttention(d_model=d_model, n_heads=n_heads, dropout=dropout)
        self.norm3 = nn.LayerNorm(d_model)
        self.ffn = FeedForward(d_model=d_model, d_ff=d_ff, dropout=dropout)
        self.dropout = nn.Dropout(dropout)

    def forward(self, x, memory):
        nx = self.norm1(x)
        x = x + self.dropout(self.self_attn(nx, nx, is_causal=True))
        x = x + self.dropout(self.cross_attn(self.norm2(x), memory))
        x = x + self.ffn(self.norm3(x))
        return x


class TransformerDecoderPreLN(nn.Module):
    def __init__(self, d_model=768, n_heads=8, d_ff=3072, n_layers=6, max_len=256, dropout=0.1):
        super().__init__()
        self.layers = nn.ModuleList([TransformerDecoderLayerPreLN(d_model=d_model, n_heads=n_heads, d_ff=d_ff, dropout=dropout) for _ in range(n_layers)])
        self.register_buffer("causal_mask", torch.triu(torch.full((max_len, max_len), float("-inf")), diagonal=1), persistent=False)

    def forward(self, x, memory):
        for layer in self.layers:
            x = layer(x, memory)
        return x


class TransformerModelPreLN(nn.Module):
    def __init__(self, vocab_size, max_src_len, max_tgt_len, d_model=768, n_heads=8, d_ff=3072, n_layers=6, dropout=0.1):
        super().__init__()
        self.vocab_size = vocab_size
        self.d_model = d_model
        self.max_src_len = max_src_len
        self.max_tgt_len = max_tgt_len
        self.token_embedding = nn.Embedding(vocab_size, d_model)
        self.register_buffer("src_pos_encoding", make_sinusoidal_table(max_src_len, d_model), persistent=False)
        self.register_buffer("tgt_pos_encoding", make_sinusoidal_table(max_tgt_len, d_model), persistent=False)
        self.dropout = nn.Dropout(dropout)
        self.encoder = TransformerEncoderPreLN(d_model=d_model, n_heads=n_heads, d_ff=d_ff, n_layers=n_layers, dropout=dropout)
        self.decoder = TransformerDecoderPreLN(d_model=d_model, n_heads=n_heads, d_ff=d_ff, n_layers=n_layers, max_len=max_tgt_len, dropout=dropout)
        self.output_projection = nn.Linear(d_model, vocab_size, bias=False)
        self.output_projection.weight = self.token_embedding.weight
        self._init_weights()

    def _init_weights(self):
        nn.init.normal_(self.token_embedding.weight, mean=0.0, std=0.02)
        for module in self.modules():
            if isinstance(module, nn.Linear):
                if module.weight is not self.token_embedding.weight:
                    nn.init.xavier_uniform_(module.weight)
                if module.bias is not None:
                    nn.init.zeros_(module.bias)
            elif isinstance(module, nn.LayerNorm):
                nn.init.ones_(module.weight); nn.init.zeros_(module.bias)

    def encode(self, src):
        _, T = src.shape
        assert T <= self.max_src_len
        x = self.token_embedding(src) * math.sqrt(self.d_model)
        x = x + self.src_pos_encoding[:T].unsqueeze(0)
        return self.encoder(self.dropout(x))

    def decode(self, tgt_in, memory):
        _, T = tgt_in.shape
        assert T <= self.max_tgt_len
        x = self.token_embedding(tgt_in) * math.sqrt(self.d_model)
        x = x + self.tgt_pos_encoding[:T].unsqueeze(0)
        return self.output_projection(self.decoder(self.dropout(x), memory))

    def forward(self, src, tgt_in):
        return self.decode(tgt_in, self.encode(src))


def build_model(vocab_size, src_len, tgt_len):
    assert src_len + tgt_len == 1024
    return TransformerModelPreLN(vocab_size=vocab_size, max_src_len=src_len, max_tgt_len=tgt_len,
                                 d_model=768, n_heads=8, d_ff=3072, n_layers=6, dropout=0.1)
'''

# ---- ARCH3 (this experiment uses Arch3 only) ----
_ARCH3_SRC = r'''def sinusoidal_pos_emb(T, d_model, device):
    pos = torch.arange(T, device=device, dtype=torch.float32).unsqueeze(1)
    div = torch.exp(torch.arange(0, d_model, 2, device=device, dtype=torch.float32) * (-math.log(10000.0) / d_model))
    pe = torch.zeros(T, d_model, device=device, dtype=torch.float32)
    pe[:, 0::2] = torch.sin(pos * div)
    pe[:, 1::2] = torch.cos(pos * div)
    return pe.unsqueeze(0)


class SlidingWindowCausalSelfAttention(nn.Module):
    def __init__(self, d_model, n_heads, block_size, window_size):
        super().__init__()
        assert d_model % n_heads == 0
        self.d_model = d_model
        self.n_heads = n_heads
        self.head_dim = d_model // n_heads
        self.window_size = window_size
        self.q_proj = nn.Linear(d_model, d_model)
        self.k_proj = nn.Linear(d_model, d_model)
        self.v_proj = nn.Linear(d_model, d_model)
        self.out_proj = nn.Linear(d_model, d_model)
        idx = torch.arange(block_size)
        causal = idx[None, :] <= idx[:, None]
        window = (idx[:, None] - idx[None, :]) < window_size
        mask = causal & window
        self.register_buffer("swa_mask", mask.view(1, 1, block_size, block_size), persistent=False)

    def forward(self, x):
        B, T, C = x.shape
        q = self.q_proj(x); k = self.k_proj(x); v = self.v_proj(x)
        q = q.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        k = k.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        v = v.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        attn_mask = self.swa_mask[:, :, :T, :T]
        y = F.scaled_dot_product_attention(q, k, v, attn_mask=attn_mask, dropout_p=0.0, is_causal=False)
        y = y.transpose(1, 2).contiguous().view(B, T, C)
        return self.out_proj(y)


class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()
        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)

    def forward(self, x):
        return self.fc2(F.gelu(self.fc1(x)))


class TransformerBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, block_size, window_size):
        super().__init__()
        self.norm1 = nn.LayerNorm(d_model)
        self.attn = SlidingWindowCausalSelfAttention(d_model=d_model, n_heads=n_heads, block_size=block_size, window_size=window_size)
        self.norm2 = nn.LayerNorm(d_model)
        self.mlp = FeedForward(d_model, d_ff)

    def forward(self, x):
        x = x + self.attn(self.norm1(x))
        x = x + self.mlp(self.norm2(x))
        return x


class LegalDecoderLM(nn.Module):
    def __init__(self, vocab_size, block_size, d_model, n_layers, n_heads, d_ff, window_size):
        super().__init__()
        self.vocab_size = vocab_size
        self.block_size = block_size
        self.d_model = d_model
        self.n_layers = n_layers
        self.token_emb = nn.Embedding(vocab_size, d_model)
        self.embed_scale = math.sqrt(d_model)
        pos = sinusoidal_pos_emb(block_size, d_model, device="cpu")
        self.register_buffer("pos_emb", pos, persistent=False)
        self.blocks = nn.ModuleList([TransformerBlock(d_model=d_model, n_heads=n_heads, d_ff=d_ff, block_size=block_size, window_size=window_size) for _ in range(n_layers)])
        self.final_norm = nn.LayerNorm(d_model)
        self.lm_head = nn.Linear(d_model, vocab_size, bias=False)
        self.lm_head.weight = self.token_emb.weight
        self.apply(self._init_weights)
        for block in self.blocks:
            block.attn.out_proj.weight.data.mul_(1.0 / math.sqrt(2 * self.n_layers))
            block.mlp.fc2.weight.data.mul_(1.0 / math.sqrt(2 * self.n_layers))

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                nn.init.zeros_(module.bias)
        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
        elif isinstance(module, nn.LayerNorm):
            nn.init.ones_(module.weight); nn.init.zeros_(module.bias)

    def forward(self, x):
        B, T = x.shape
        assert T <= self.block_size, f"Sequence length {T} exceeds block_size {self.block_size}"
        tok = self.token_emb(x) * self.embed_scale
        pos = self.pos_emb[:, :T, :].to(tok.dtype)
        h = tok + pos
        for block in self.blocks:
            h = block(h)
        h = self.final_norm(h)
        return self.lm_head(h)


def build_model(vocab_size, block_size, window_size=256):
    return LegalDecoderLM(vocab_size=vocab_size, block_size=block_size,
                          d_model=1024, n_layers=16, n_heads=16, d_ff=4096, window_size=window_size)
'''


def _namespace(src, name):
    ns = {"math": math, "torch": torch, "nn": nn, "F": F, "__name__": f"phase1_{name}"}
    exec(compile(src, f"<phase1 {name} definition>", "exec"), ns)
    return ns


def _strip_compile_prefix(sd):
    return {(k[len("_orig_mod."):] if k.startswith("_orig_mod.") else k): v for k, v in sd.items()}


def load_backbone(name, ckpt_path=None, device="cuda"):
    from types import SimpleNamespace
    ckpt_path = ckpt_path or os.environ.get(f"CKPT_{name.upper()}") or DEFAULT_CKPTS[name]
    if name == "arch1":
        ns = _namespace(_ARCH1_SRC, name)
        model, max_len, d_model = ns["build_model"](VOCAB, 1024), 1024, 768
        def hidden(ids):
            B, T = ids.shape
            h = model.token_emb(ids) + model.pos_emb(torch.arange(T, device=ids.device))
            for block in model.blocks: h = block(h)
            return model.final_norm(h)
    elif name == "arch3":
        ns = _namespace(_ARCH3_SRC, name)
        model, max_len, d_model = ns["build_model"](VOCAB, 1024, 256), 1024, 1024
        def hidden(ids):
            B, T = ids.shape
            h = model.token_emb(ids) * model.embed_scale + model.pos_emb[:, :T, :].to(model.token_emb.weight.dtype)
            for block in model.blocks: h = block(h)
            return model.final_norm(h)
    elif name == "arch2":
        ns = _namespace(_ARCH2_SRC, name)
        model, max_len, d_model = ns["build_model"](VOCAB, 768, 256), 768, 768
        def hidden(ids):
            return model.encode(ids)
    else:
        raise ValueError("name must be arch1, arch2 or arch3")

    ck = torch.load(ckpt_path, map_location="cpu", weights_only=False)
    state = ck["model"] if isinstance(ck, dict) and "model" in ck else ck
    model.load_state_dict(_strip_compile_prefix(state), strict=True)
    meta = {"name": name, "ckpt": ckpt_path,
            "n_params": sum(p.numel() for p in model.parameters()),
            "step": ck.get("step") if isinstance(ck, dict) else None,
            "best": ck.get("best") if isinstance(ck, dict) else None}
    model = model.to(device).eval()
    return SimpleNamespace(model=model, hidden=hidden, max_len=max_len, d_model=d_model, meta=meta)


# =============================================================================
# LOCATE CHECKPOINT + TOKENIZER, LOAD DATASET
# =============================================================================
INPUT_ROOT = Path("/kaggle/input")

arch3_default = Path(os.environ.get("CKPT_ARCH3", DEFAULT_CKPTS["arch3"]))
if arch3_default.is_file():
    ARCH3_CKPT = arch3_default
else:
    candidates = sorted((p for p in INPUT_ROOT.rglob("best.pt") if "arch3" in str(p).lower()),
                        key=lambda p: ("arch3-model" not in str(p).lower(), len(str(p))))
    if not candidates:
        raise FileNotFoundError("Arch3 Phase-1 best.pt not found. Attach the Arch3 model dataset.")
    ARCH3_CKPT = candidates[0]

preferred_tokenizer = Path("/kaggle/input/datasets/achiver07/phase1-tokens-v40k-clean1/tokenizer.json")
if preferred_tokenizer.is_file():
    TOKENIZER_PATH = preferred_tokenizer
else:
    toks = sorted(INPUT_ROOT.rglob("tokenizer.json"),
                  key=lambda p: ("phase1-tokens-v40k-clean1" not in str(p).lower(), len(str(p))))
    if not toks:
        raise FileNotFoundError("Phase-1 tokenizer.json not found. Attach the Phase-1 tokenizer dataset.")
    TOKENIZER_PATH = toks[0]

if not all(p.is_file() for p in (TRAIN_PATH, DEV_PATH, TEST_PATH)):
    print("Phase-4 files missing; attempting HF download...")
    try:
        from kaggle_secrets import UserSecretsClient
        from huggingface_hub import snapshot_download
        hf_token = UserSecretsClient().get_secret("HF_TOKEN")
        snapshot_download(repo_id="Exploration-Lab/CS-781-Capstone", repo_type="dataset",
                          token=hf_token, allow_patterns=["Phase_4_CJPE/**"],
                          local_dir="/kaggle/working/Phase_4_CJPE")
    except Exception as exc:
        raise RuntimeError("HF download failed. Check internet, HF_TOKEN secret, and repo access.") from exc

for p in (TRAIN_PATH, DEV_PATH, TEST_PATH):
    if not p.is_file():
        raise FileNotFoundError(f"CJPE file missing after download: {p}")

print("Arch3 checkpoint:", ARCH3_CKPT)
print("Tokenizer:", TOKENIZER_PATH)


def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)


train_split_records = load_json(TRAIN_PATH)
dev_split_records = load_json(DEV_PATH)
test_raw = load_json(TEST_PATH)
test_records = [{"id": row["id"], "text": row.get("text", "")} for row in test_raw]
del test_raw

for name, records in [("train", train_split_records), ("dev", dev_split_records)]:
    assert records, f"{name} split is empty"
    assert all("id" in r and "text" in r for r in records), f"Malformed {name} example"
    assert all(r.get("label") in (0, 1) for r in records), f"Invalid/missing labels in {name}"
assert test_records, "test split is empty"
assert all("id" in r and "text" in r for r in test_records), "Malformed test example"

test_ids = [str(r["id"]) for r in test_records]
assert len(set(test_ids)) == len(test_ids), "Duplicate test IDs"

SOURCE_TRAIN_COUNT = len(train_split_records)
SOURCE_DEV_COUNT = len(dev_split_records)
corpus_records = train_split_records + dev_split_records

if MAX_TRAIN_DOCS is not None:
    rng = np.random.default_rng(SEED)
    n_keep = min(int(MAX_TRAIN_DOCS), len(corpus_records))
    chosen = np.sort(rng.choice(len(corpus_records), size=n_keep, replace=False))
    corpus_records = [corpus_records[i] for i in chosen]

train_records = corpus_records
y_train_full = np.asarray([int(r["label"]) for r in train_records], dtype=np.int64)
print(f"Full labelled corpus: {len(train_records):,}")
print(f"Test examples: {len(test_records):,}")
print(f"Corpus labels: {Counter(y_train_full.tolist())}")

# ------------------------------------------------------------------ Holdout
if HOLDOUT_FRAC > 0:
    _rng = np.random.default_rng(SEED)
    _perm = _rng.permutation(len(train_records))
    _n_ho = max(1, int(HOLDOUT_FRAC * len(train_records)))
    ho_idx = _perm[:_n_ho]
    tr_idx = _perm[_n_ho:]
    train_records_fit = [train_records[i] for i in tr_idx]
    holdout_records = [train_records[i] for i in ho_idx]
    y_ho = y_train_full[ho_idx]
    print(f"Self-created holdout (for threshold tuning, NOT dev): "
          f"{len(holdout_records):,}  | fit: {len(train_records_fit):,}")
else:
    tr_idx = np.arange(len(train_records))
    ho_idx = np.array([], dtype=int)
    train_records_fit = train_records
    holdout_records = []
    y_ho = np.array([], dtype=np.int64)
    print("HOLDOUT_FRAC=0 → strict fixed-threshold policy, no holdout created.")

y_train = y_train_full[tr_idx]

# =============================================================================
# TOKENIZATION  (full document first, then zone selection; cached)
# =============================================================================
tokenizer = Tokenizer.from_file(str(TOKENIZER_PATH))
BOS_ID, EOS_ID, PAD_ID = 1, 2, 0
N_ZONES = 4


def select_token_zones(full_token_ids, zone_tokens):
    ids = list(map(int, full_token_ids))
    n = len(ids)
    if n == 0:
        return [[], [], [], []]
    if n <= N_ZONES * zone_tokens:
        b = np.linspace(0, n, N_ZONES + 1, dtype=int)
        return [ids[b[i]:b[i + 1]] for i in range(N_ZONES)]
    last_start = n - zone_tokens
    starts = [0,
              int(round(last_start / 3)),
              int(round(2 * last_start / 3)),
              last_start]
    starts = [max(0, min(s, last_start)) for s in starts]
    starts[-1] = last_start
    return [ids[s:s + zone_tokens] for s in starts]


def encode_records(records, split_name, batch_docs=64):
    n = len(records)
    content_len = MAX_ZONE_LEN - 2
    token_ids = np.zeros((n, N_ZONES, MAX_ZONE_LEN), dtype=np.int32)
    lengths = np.zeros((n, N_ZONES), dtype=np.int16)
    tokenizer.no_truncation()
    for start in tqdm(range(0, n, batch_docs), desc=f"Tokenizing {split_name}"):
        batch = records[start:start + batch_docs]
        texts = [str(row.get("text") or "") for row in batch]
        encodings = tokenizer.encode_batch(texts, add_special_tokens=False)
        for offset, enc in enumerate(encodings):
            doc_i = start + offset
            zones = select_token_zones(enc.ids, content_len)
            for zi, zone in enumerate(zones):
                content = [int(t) for t in zone[:content_len]]
                seq = [BOS_ID] + content + [EOS_ID]
                L = min(len(seq), MAX_ZONE_LEN)
                token_ids[doc_i, zi, :L] = seq[:L]
                lengths[doc_i, zi] = L
    assert token_ids.min() >= 0 and token_ids.max() < VOCAB
    return token_ids, lengths


def _cache_path(tag):
    key = f"{tag}_n{len(train_records)}_t{len(test_records)}_z{N_ZONES}_L{MAX_ZONE_LEN}_ho{HOLDOUT_FRAC}"
    return OUT_DIR / f"token_cache_{key}.npz"


def _get_tokens(records, split_name, cache_tag):
    cache = _cache_path(cache_tag)
    if cache.is_file():
        d = np.load(cache)
        print("Loaded token cache:", cache)
        return d["X"], d["L"]
    X, L = encode_records(records, split_name, batch_docs=64)
    np.savez_compressed(cache, X=X, L=L)
    print("Wrote token cache:", cache)
    return X, L


X_train, L_train = _get_tokens(train_records_fit, "fit corpus", "fit")
X_ho, L_ho = (_get_tokens(holdout_records, "holdout", "ho") if HOLDOUT_FRAC > 0 else (None, None))
X_test, L_test = _get_tokens(test_records, "test inputs", "test")

print("Encoded shapes:",
      "train:", X_train.shape,
      "holdout:", (X_ho.shape if X_ho is not None else None),
      "test:", X_test.shape)

# =============================================================================
# LOAD ARCH3 + INJECT ADALORA
# =============================================================================
print("PEFT:", getattr(peft, "__version__", "unknown"))
bb3 = load_backbone("arch3", ckpt_path=str(ARCH3_CKPT), device=str(DEVICE))
assert bb3.meta["n_params"] == EXPECTED_PARAMS["arch3"], "Wrong Arch3 checkpoint."
print("Loaded:", bb3.meta)

TARGET_SUFFIXES = ("q_proj", "k_proj", "v_proj", "out_proj", "fc1", "fc2")
target_names = [n for n, m in bb3.model.named_modules()
                if isinstance(m, nn.Linear) and n.split(".")[-1] in TARGET_SUFFIXES]
assert len(target_names) == 16 * 6, f"Expected 96 targets, found {len(target_names)}"
print("AdaLoRA targets:", len(target_names))

# Loader: shuffle, drop_last keeps shapes fixed for torch.compile; num_workers overlaps CPU with GPU.
train_loader = DataLoader(
    TensorDataset(torch.from_numpy(X_train).long(),
                  torch.from_numpy(L_train).long(),
                  torch.from_numpy(y_train).long()),
    batch_size=BATCH_DOCS, shuffle=True,
    num_workers=NUM_WORKERS, pin_memory=True, drop_last=True,
)
holdout_loader = (DataLoader(
    TensorDataset(torch.from_numpy(X_ho).long(), torch.from_numpy(L_ho).long()),
    batch_size=BATCH_DOCS, shuffle=False,
    num_workers=NUM_WORKERS, pin_memory=True,
) if HOLDOUT_FRAC > 0 else None)

steps_per_epoch = len(train_loader)
total_steps = EPOCHS * steps_per_epoch
assert total_steps >= 8, f"Too few steps: {total_steps}"
TINIT = max(1, int(total_steps * 0.05))
TFINAL = max(1, int(total_steps * 0.10))
if TINIT >= total_steps - TFINAL:
    TINIT, TFINAL = 0, max(1, total_steps // 5)

DELTA_T = 8   # CHANGED: 25 → 8  (batch 16 has ~4× fewer steps than batch 4)
adalora_config = AdaLoraConfig(
    init_r=16,                 # CHANGED: 8 → 16
    target_r=8,                # CHANGED: 4 → 8
    tinit=TINIT, tfinal=TFINAL, deltaT=DELTA_T,
    beta1=0.85, beta2=0.85,
    lora_alpha=16, lora_dropout=0.05, orth_reg_weight=0.1,
    total_step=total_steps,
    target_modules=list(TARGET_SUFFIXES),
    task_type=None, bias="none",
)
adalora_model = get_peft_model(bb3.model, adalora_config)
adalora_model.train()
adalora_model.print_trainable_parameters()

# =============================================================================
# CLASSIFIER HEAD  (learned cross-zone Transformer, replaces hand-picked concat)
# =============================================================================
class CJPEZoneClassifier(nn.Module):
    def __init__(self, d_model=1024, n_zones=4, dropout=0.25, n_heads=8, n_layers=2):
        super().__init__()
        self.n_zones = n_zones
        self.zone_pos = nn.Parameter(torch.zeros(n_zones, d_model))
        enc = nn.TransformerEncoderLayer(
            d_model=d_model, nhead=n_heads, dim_feedforward=2048,
            dropout=dropout, activation="gelu", batch_first=True, norm_first=True,
        )
        self.cross_zone = nn.TransformerEncoder(enc, num_layers=n_layers)
        self.zone_score = nn.Sequential(
            nn.LayerNorm(d_model), nn.Linear(d_model, 128), nn.Tanh(), nn.Linear(128, 1)
        )
        in_dim = d_model * 4 + n_zones
        self.classifier = nn.Sequential(
            nn.LayerNorm(in_dim), nn.Dropout(dropout),
            nn.Linear(in_dim, 512), nn.GELU(), nn.Dropout(dropout),
            nn.Linear(512, 2),
        )

    def forward(self, zone_reps):
        x = zone_reps + self.zone_pos.unsqueeze(0)              # [B, 4, D]
        x = self.cross_zone(x)
        w = torch.softmax(self.zone_score(x).squeeze(-1), dim=1)  # [B, 4]
        pooled = (x * w.unsqueeze(-1)).sum(dim=1)
        fused = torch.cat([pooled, x.max(dim=1).values, x.mean(dim=1),
                           x[:, -1], w], dim=-1)
        return self.classifier(fused), w


classifier = CJPEZoneClassifier(d_model=bb3.d_model).to(DEVICE)


# =============================================================================
# HIDDEN + POOLING  (compile only the training-side forward)
# =============================================================================
def arch3_hidden(ids):
    model = bb3.model
    B, T = ids.shape
    if T > bb3.max_len:
        raise ValueError(f"Input length {T} exceeds Arch3 max ({bb3.max_len})")
    h = model.token_emb(ids) * model.embed_scale + model.pos_emb[:, :T, :].to(model.token_emb.weight.dtype)
    for block in model.blocks:
        if USE_GRAD_CHECKPOINTING and model.training and torch.is_grad_enabled():
            h = checkpoint(block, h, use_reentrant=False)
        else:
            h = block(h)
    return model.final_norm(h)


if USE_GRAD_CHECKPOINTING:
    print("NOTE: gradient checkpointing is on; torch.compile is skipped.")
else:
    # fullgraph=False tolerates graph breaks from AdaLoRA's dynamic rank masks.
    arch3_hidden = torch.compile(arch3_hidden, mode="default", fullgraph=False)
    print("torch.compile active on training forward pass.")


def zone_representations(input_ids, lengths):
    B, Z, T = input_ids.shape
    flat_ids = input_ids.reshape(B * Z, T)
    flat_len = lengths.reshape(B * Z).clamp(min=1, max=T)
    h = arch3_hidden(flat_ids)
    positions = torch.arange(T, device=h.device).unsqueeze(0)
    mask = (positions < flat_len.unsqueeze(1)).unsqueeze(-1).to(h.dtype)
    mean_pool = (h * mask).sum(dim=1) / mask.sum(dim=1).clamp(min=1.0)
    row = torch.arange(B * Z, device=h.device)
    end_pool = h[row, flat_len - 1]
    return (0.5 * mean_pool + 0.5 * end_pool).reshape(B, Z, -1)


# Identity / shape smoke test using FULL training shape (also warms compile):
adalora_model.eval()
with torch.no_grad(), torch.amp.autocast("cuda", dtype=torch.float16):
    _warm_ids = torch.from_numpy(X_train[:BATCH_DOCS]).long().to(DEVICE).reshape(-1, MAX_ZONE_LEN)
    _warm_len = torch.from_numpy(L_train[:BATCH_DOCS]).long().to(DEVICE).reshape(-1)
    _warm_h = arch3_hidden(_warm_ids)
assert _warm_h.shape == (BATCH_DOCS * N_ZONES, MAX_ZONE_LEN, bb3.d_model)
print("Compile warm-up done; hidden shape:", tuple(_warm_h.shape))
del _warm_ids, _warm_len, _warm_h
adalora_model.train()

# =============================================================================
# OPTIMIZER / SCHEDULER / LOSS
# =============================================================================
adapter_params = [p for p in adalora_model.parameters() if p.requires_grad]
head_params = list(classifier.parameters())
trainable_params = adapter_params + head_params

n_base_total = int(bb3.meta["n_params"])
n_trainable = sum(p.numel() for p in trainable_params)

# CHANGED: adapter 8e-5→1.5e-4, head 2e-4→3e-4 (3-epoch cosine can take a stronger LR).
optimizer = torch.optim.AdamW(
    [{"params": adapter_params, "lr": 1.5e-4},
     {"params": head_params,    "lr": 3.0e-4}],
    weight_decay=0.01,
)


def _cos_warmup(step, total, warm_frac=0.05, min_frac=0.05):
    w = max(1, int(total * warm_frac))
    if step < w:
        return step / w
    p = (step - w) / max(1, total - w)
    return min_frac + (1 - min_frac) * 0.5 * (1.0 + math.cos(math.pi * p))


scheduler = torch.optim.lr_scheduler.LambdaLR(
    optimizer, lr_lambda=lambda s: _cos_warmup(s, total_steps)
)

# Class weights from the actual fit split:
counts = np.bincount(y_train, minlength=2).astype(np.float64)
cw = (len(y_train) / np.maximum(counts, 1.0)) ** 0.5
cw = cw / cw.mean()
criterion = nn.CrossEntropyLoss(
    weight=torch.tensor(cw, dtype=torch.float32, device=DEVICE), label_smoothing=0.02
)
scaler = torch.amp.GradScaler("cuda", enabled=True)

print(f"Total steps={total_steps:,} | tinit={TINIT:,} | tfinal={TFINAL:,} | deltaT={DELTA_T}")
print(f"Backbone params={n_base_total:,} | trainable (adapters + head)={n_trainable:,} "
      f"({100*n_trainable/(n_base_total+n_trainable):.3f}%)")

# =============================================================================
# TRAINING LOOP
# =============================================================================
torch.cuda.reset_peak_memory_stats()
start_time = time.time()
epoch_losses = []

if ADAPTER_DIR.exists():
    shutil.rmtree(ADAPTER_DIR)
ADAPTER_DIR.mkdir(parents=True, exist_ok=True)

global_step = 0
for epoch in range(1, EPOCHS + 1):
    adalora_model.train(); classifier.train()
    running_loss = 0.0
    skipped = 0
    pbar = tqdm(enumerate(train_loader, start=1), total=len(train_loader),
                desc=f"Epoch {epoch}/{EPOCHS}")
    for batch_idx, (ids, lens, labels) in pbar:
        ids = ids.to(DEVICE, non_blocking=True)
        lens = lens.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        optimizer.zero_grad(set_to_none=True)

        with torch.amp.autocast("cuda", dtype=torch.float16):
            z = zone_representations(ids, lens)
            logits, _ = classifier(z)
            loss = criterion(logits.float(), labels)

        if not torch.isfinite(loss):
            skipped += 1
            optimizer.zero_grad(set_to_none=True)
            continue

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(trainable_params, max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()
        scheduler.step()                          # CHANGED: cosine update
        adalora_model.base_model.update_and_allocate(global_step)
        global_step += 1

        running_loss += float(loss.detach().cpu())
        if batch_idx % 20 == 0:
            pbar.set_postfix(loss=f"{running_loss / max(1, batch_idx - skipped):.4f}",
                             lr=f"{optimizer.param_groups[0]['lr']:.2e}",
                             step=global_step)

    mean_train_loss = float(running_loss / max(1, len(train_loader) - skipped))
    epoch_losses.append({"epoch": epoch, "mean_train_loss": mean_train_loss,
                         "skipped_batches": int(skipped)})
    print(f"Epoch {epoch}: train_loss={mean_train_loss:.6f} | skipped={skipped}")

    adalora_model.save_pretrained(str(ADAPTER_DIR), safe_serialization=True)
    torch.save(classifier.state_dict(), HEAD_PATH)

train_seconds = time.time() - start_time
peak_gpu_gb = torch.cuda.max_memory_allocated() / 1024**3
print(f"Training done in {train_seconds/60:.1f} min | peak VRAM={peak_gpu_gb:.2f} GiB")

# =============================================================================
# THRESHOLD TUNING ON SELF-CREATED HOLDOUT  (NOT dev)
# =============================================================================
adalora_model.eval(); classifier.eval()


@torch.no_grad()
def _predict_loader(loader, desc):
    probs, attn = [], []
    for ids, lens in tqdm(loader, desc=desc):
        ids = ids.to(DEVICE, non_blocking=True)
        lens = lens.to(DEVICE, non_blocking=True)
        with torch.amp.autocast("cuda", dtype=torch.float16):
            z = zone_representations(ids, lens)
            logits, w = classifier(z)
        probs.extend(torch.softmax(logits.float(), dim=-1)[:, 1].cpu().numpy().tolist())
        attn.append(w.float().cpu().numpy())
    return np.asarray(probs), np.concatenate(attn, axis=0)


TUNED_THRESHOLD = 0.5
HOLDOUT_F1 = None
if HOLDOUT_FRAC > 0 and len(y_ho) > 0:
    p_ho, _ = _predict_loader(holdout_loader, "Holdout (self-created, NOT dev)")
    f1s = [f1_score(y_ho, (p_ho >= t).astype(int), pos_label=1, zero_division=0)
           for t in THRESHOLD_GRID]
    best_i = int(np.argmax(f1s))
    TUNED_THRESHOLD = float(THRESHOLD_GRID[best_i])
    HOLDOUT_F1 = float(f1s[best_i])
    print(f"Holdout F1 (self-created, NOT dev) = {HOLDOUT_F1:.4f} "
          f"at threshold {TUNED_THRESHOLD:.3f}")
else:
    print("No holdout → using fixed threshold 0.5.")

# =============================================================================
# FREE + RELOAD SAVED WEIGHTS FOR TEST INFERENCE
# =============================================================================
for _name in ("adalora_model", "bb3", "classifier", "optimizer", "scheduler",
              "scaler", "trainable_params", "adapter_params", "head_params"):
    globals().pop(_name, None)
if torch.cuda.is_available():
    torch.cuda.empty_cache()
gc.collect()

bb3 = load_backbone("arch3", ckpt_path=str(ARCH3_CKPT), device=str(DEVICE))
adalora_model = PeftModel.from_pretrained(bb3.model, str(ADAPTER_DIR), is_trainable=False)
adalora_model.eval()
classifier = CJPEZoneClassifier(d_model=bb3.d_model).to(DEVICE)
classifier.load_state_dict(torch.load(HEAD_PATH, map_location=DEVICE, weights_only=True))
classifier.eval()


# Uncompiled inference forward (compile is off at inference — not the bottleneck).
def arch3_hidden(ids):
    model = bb3.model
    B, T = ids.shape
    h = model.token_emb(ids) * model.embed_scale + model.pos_emb[:, :T, :].to(model.token_emb.weight.dtype)
    for block in model.blocks:
        h = block(h)
    return model.final_norm(h)


test_loader = DataLoader(
    TensorDataset(torch.from_numpy(X_test).long(), torch.from_numpy(L_test).long()),
    batch_size=BATCH_DOCS, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True,
)
p_test, attn_test = _predict_loader(test_loader, "Test prediction")

preds = (p_test >= TUNED_THRESHOLD).astype(np.int64)

submission = [{"id": str(row["id"]), "pred": int(pred)}
              for row, pred in zip(test_records, preds)]
assert len(submission) == len(test_records)
assert [x["id"] for x in submission] == test_ids
assert len({x["id"] for x in submission}) == len(submission)
assert all(type(x["pred"]) is int and x["pred"] in (0, 1) for x in submission)

# =============================================================================
# SAVE OUTPUTS + METADATA
# =============================================================================
training_metadata = {
    "experiment": f"Phase 4 Exp2 — Arch3 + AdaLoRA FAST+BEST "
                  f"(batch={BATCH_DOCS}, compile=True, epochs={EPOCHS}, holdout={HOLDOUT_FRAC})",
    "architecture": "Phase 1 Arch3 (242.5M parameters)",
    "adaptation": "AdaLoRA (dynamic rank allocation)",
    "checkpoint": str(ARCH3_CKPT),
    "tokenizer": str(TOKENIZER_PATH),
    "original_train_examples": int(SOURCE_TRAIN_COUNT),
    "development_examples_merged_into_training": int(SOURCE_DEV_COUNT),
    "full_corpus_training_examples": int(len(train_records)),
    "fit_examples": int(len(train_records_fit)),
    "holdout_examples_self_created_not_dev": int(len(holdout_records)),
    "test_examples": int(len(test_records)),
    "epochs": int(EPOCHS),
    "batch_docs": int(BATCH_DOCS),
    "num_workers": int(NUM_WORKERS),
    "drop_last": True,
    "torch_compile": bool(not USE_GRAD_CHECKPOINTING),
    "zone_length": int(MAX_ZONE_LEN),
    "target_modules": list(TARGET_SUFFIXES),
    "init_rank": 16,
    "target_rank": 8,
    "tinit": int(TINIT),
    "tfinal": int(TFINAL),
    "deltaT": int(DELTA_T),
    "lr_adapter": 1.5e-4,
    "lr_head": 3.0e-4,
    "scheduler": "cosine_with_5pct_warmup",
    "holdout_frac": float(HOLDOUT_FRAC),
    "tuned_threshold": float(TUNED_THRESHOLD),
    "holdout_f1_self_created_not_dev": (float(HOLDOUT_F1) if HOLDOUT_F1 is not None else None),
    "epoch_training_log": epoch_losses,
    "training_seconds": float(train_seconds),
    "peak_gpu_memory_gib": float(peak_gpu_gb),
    "mean_zone_attention_test": attn_test.mean(axis=0).tolist(),
    "test_positive_predictions": int((preds == 1).sum()),
    "test_negative_predictions": int((preds == 0).sum()),
    "test_labels_used": False,
    "phase1_base_parameters": int(n_base_total),
    "trainable_parameters_adalora_plus_head": int(n_trainable),
    "trainable_parameter_percent": float(100 * n_trainable / (n_base_total + n_trainable)),
}

with open(SUBMISSION_PATH, "w", encoding="utf-8") as f:
    json.dump(submission, f, ensure_ascii=False, indent=2)
with open(RESULTS_PATH, "w", encoding="utf-8") as f:
    json.dump(training_metadata, f, ensure_ascii=False, indent=2)
with zipfile.ZipFile(ZIP_PATH, "w", compression=zipfile.ZIP_DEFLATED) as zf:
    zf.write(SUBMISSION_PATH, arcname="prediction.json")

try:
    cfg = adalora_model.peft_config.get("default")
    rank_pattern = getattr(cfg, "rank_pattern", None) if cfg is not None else None
    rank_json = {str(k): v for k, v in rank_pattern.items()} if rank_pattern else {}
    with open(OUT_DIR / "adalora_rank_pattern.json", "w", encoding="utf-8") as f:
        json.dump(rank_json, f, indent=2)
    print("Saved rank pattern entries:", len(rank_json))
except Exception as exc:
    print("Could not serialize rank_pattern:", repr(exc))

# Format-only validation
with open(SUBMISSION_PATH, "r", encoding="utf-8") as f:
    check = json.load(f)
assert isinstance(check, list) and len(check) == len(test_ids)
assert [str(x["id"]) for x in check] == test_ids
assert all(set(x.keys()) == {"id", "pred"} for x in check)
assert all(type(x["pred"]) is int and x["pred"] in (0, 1) for x in check)
with zipfile.ZipFile(ZIP_PATH, "r") as zf:
    assert zf.namelist() == ["prediction.json"]

print("\n" + "=" * 72)
print("PHASE 4 EXP2 — ARCH3 + ADALORA (FAST + BEST-RESULT VARIANT)")
print("=" * 72)
print(f"Fit examples:                         {len(train_records_fit):,}")
print(f"Holdout (self-created, NOT dev):      {len(holdout_records):,}")
print(f"Test predictions:                     {len(submission):,}")
print(f"Tuned threshold:                      {TUNED_THRESHOLD:.3f}")
if HOLDOUT_F1 is not None:
    print(f"Holdout F1 (self-created, NOT dev):   {HOLDOUT_F1:.4f}")
print(f"Training time:                        {train_seconds/60:.1f} min")
print(f"Peak VRAM:                            {peak_gpu_gb:.2f} GiB")
print("Test prediction counts:", Counter(int(x["pred"]) for x in submission))
print("Submission JSON:", SUBMISSION_PATH)
print("Submission ZIP:", ZIP_PATH)
print("Metadata:", RESULTS_PATH)
print("Format validation: PASS (no test labels used)")
print("\nOutput folder contents:")
for p in sorted(OUT_DIR.rglob("*")):
    if p.is_file():
        print(f"  {p} ({p.stat().st_size / 1024:.1f} KiB)")
```

```python
# PHASE 4 — EXPERIMENT 2: ARCH3 + ADALORA (ONE-CELL, IMPROVED TOKEN ZONES)
# Full-corpus training: train.json + dev.json; NO development evaluation or threshold tuning.
# Improvement: tokenize each complete legal document first, then select four token windows,
# with the last window anchored to the actual document end. Zone length is 256 tokens.
# Output: /kaggle/working/Phase4_Submissions/Exp2_Arch3_AdaLoRA_Improved/
# Requires Kaggle GPU and the attached Phase 1 Arch3 checkpoint + Phase 1 tokenizer.

import os, sys, json, time, math, random, gc, zipfile, shutil, warnings
from pathlib import Path
from collections import Counter

import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset
from torch.utils.checkpoint import checkpoint
from tokenizers import Tokenizer
from sklearn.metrics import f1_score, precision_score, recall_score, confusion_matrix, classification_report
from tqdm.auto import tqdm

try:
    import peft
    from peft import AdaLoraConfig, get_peft_model, PeftModel
except Exception:
    import subprocess
    print("PEFT not found; attempting to install peft==0.17.1 (library only, no model weights).")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "peft==0.17.1"])
    import peft
    from peft import AdaLoraConfig, get_peft_model, PeftModel

assert torch.cuda.is_available(), "Select a GPU accelerator in Kaggle Settings before running this notebook."
DEVICE = torch.device("cuda")
SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)

DATA_ROOT = Path("/kaggle/working/Phase_4_CJPE/Phase_4_CJPE")
TRAIN_PATH = DATA_ROOT / "Train" / "train.json"
DEV_PATH   = DATA_ROOT / "Train" / "dev.json"
TEST_PATH  = DATA_ROOT / "Test" / "test.json"

OUT_DIR = Path("/kaggle/working/Phase4_Submissions/Exp2_Arch3_AdaLoRA_Improved")
OUT_DIR.mkdir(parents=True, exist_ok=True)
ADAPTER_DIR = OUT_DIR / "adalora_adapter"
HEAD_PATH = OUT_DIR / "classifier_head.pt"
SUBMISSION_PATH = OUT_DIR / "prediction.json"
ZIP_PATH = OUT_DIR / "submission.zip"
RESULTS_PATH = OUT_DIR / "training_metadata.json"

# Experiment controls. Set MAX_TRAIN_DOCS to an integer only for a quick smoke test.
MAX_TRAIN_DOCS = None       # None = use all labelled train + dev examples (39,375 total)
EPOCHS = 1                  # final-fit default; no validation-based epoch selection
BATCH_DOCS = 2              # 4 zones x 256 tokens; reduce to 1 if CUDA OOM
MAX_ZONE_LEN = 256          # per zone including BOS/EOS; 254 content tokens
USE_GRAD_CHECKPOINTING = True

print("PyTorch:", torch.__version__)
print("GPU:", torch.cuda.get_device_name(0))
print("GPU memory (GiB):", round(torch.cuda.get_device_properties(0).total_memory / 1024**3, 2))
print("Data root:", DATA_ROOT)
print("Output folder:", OUT_DIR)

#!/usr/bin/env python3
"""
phase2_backbones.py -- load the Phase-1 Arch 1 / 2 / 3 backbones for Phase-2 NER and test them.

The three model definitions below are copied VERBATIM from your Phase-1 training scripts
(Arch 1: train_model_1_initfix.py, Arch 2: your Pre-LN script, Arch 3: train_model_3_swa_v2.py).
Each is executed in its own namespace because they reuse class names.

What this file adds (no modelling decisions, no head, no adapters):
  * load_backbone(name, ckpt_path)  -> Backbone(model, hidden(ids), max_len, d_model, meta)
        hidden(ids) returns the (B, T, d_model) states your NER head would read:
          arch1 / arch3: after final_norm, exactly what lm_head reads
          arch2        : model.encode(ids) = encoder output, RAW (this model has no final norm)
  * selftest(name, ckpt_path)       -> strict load, parameter count, checkpoint step/best,
        lm_head(hidden(x)) == model(x) parity (arch1/3), causality (arch1/3),
        bidirectionality (arch2), and activation-scale statistics.

Kaggle: paste into one cell and run; paths default to the ones you sent and can be overridden with
env vars CKPT_ARCH1 / CKPT_ARCH2 / CKPT_ARCH3. Or save as phase2_backbones.py and `import` it.
"""
import math
import os
from types import SimpleNamespace

import torch
import torch.nn as nn
import torch.nn.functional as F

VOCAB = 40000
EXPECTED_PARAMS = {"arch1": 88_210_944, "arch2": 129_957_888, "arch3": 242_501_632}

DEFAULT_CKPTS = {
    "arch1": "/kaggle/input/models/achiver7/arch1-modeldecoder/transformers/default/1/run_arch1_real/best.pt",
    "arch2": "/kaggle/input/models/achiver7/arch2-model/transformers/default/1/run_arch2_preln_efficient/best.pt",
    "arch3": "/kaggle/input/datasets/achiver7/arch3-model/run_arch3_swa_v2/best.pt",
}

# ======================= verbatim model definitions (one namespace each) =======================
_ARCH1_SRC = r'''INIT_STD = 0.02


def _init_weights(m):
    if isinstance(m, nn.Linear):
        nn.init.normal_(m.weight, mean=0.0, std=INIT_STD)
        if m.bias is not None:
            nn.init.zeros_(m.bias)
    elif isinstance(m, nn.Embedding):
        nn.init.normal_(m.weight, mean=0.0, std=INIT_STD)


class CausalSelfAttention(nn.Module):
    def __init__(self, d_model, n_heads, block_size):
        super().__init__()
        assert d_model % n_heads == 0
        self.d_model = d_model
        self.n_heads = n_heads
        self.head_dim = d_model // n_heads

        self.q_proj = nn.Linear(d_model, d_model)
        self.k_proj = nn.Linear(d_model, d_model)
        self.v_proj = nn.Linear(d_model, d_model)
        self.out_proj = nn.Linear(d_model, d_model)

        mask = torch.tril(torch.ones(block_size, block_size)).view(1, 1, block_size, block_size)
        self.register_buffer("mask", mask)

    def forward(self, x):
        B, T, C = x.shape

        q = self.q_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        k = self.k_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        v = self.v_proj(x).view(B, T, self.n_heads, self.head_dim).transpose(1, 2)

        att = (q @ k.transpose(-2, -1)) / math.sqrt(self.head_dim)
        att = att.masked_fill(self.mask[:, :, :T, :T] == 0, float("-inf"))
        att = F.softmax(att, dim=-1)

        y = att @ v
        y = y.transpose(1, 2).contiguous().view(B, T, C)
        return self.out_proj(y)


class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()
        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)

    def forward(self, x):
        return self.fc2(F.gelu(self.fc1(x)))


class TransformerBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, block_size):
        super().__init__()
        self.norm1 = nn.LayerNorm(d_model)
        self.attn = CausalSelfAttention(d_model, n_heads, block_size)
        self.norm2 = nn.LayerNorm(d_model)
        self.mlp = FeedForward(d_model, d_ff)

    def forward(self, x):
        x = x + self.attn(self.norm1(x))
        x = x + self.mlp(self.norm2(x))
        return x


class LegalDecoderLM(nn.Module):
    def __init__(self, vocab_size, block_size, d_model, n_layers, n_heads, d_ff):
        super().__init__()
        self.vocab_size = vocab_size
        self.block_size = block_size
        self.d_model = d_model

        self.token_emb = nn.Embedding(vocab_size, d_model)
        self.pos_emb = nn.Embedding(block_size, d_model)

        self.blocks = nn.ModuleList([
            TransformerBlock(d_model, n_heads, d_ff, block_size)
            for _ in range(n_layers)
        ])

        self.final_norm = nn.LayerNorm(d_model)
        self.lm_head = nn.Linear(d_model, vocab_size, bias=False)
        self.lm_head.weight = self.token_emb.weight      # tied embeddings

        self.apply(_init_weights)                        # CHANGE 1: small-std init

    def forward(self, x):
        B, T = x.shape
        assert T <= self.block_size

        tok = self.token_emb(x)
        pos = self.pos_emb(torch.arange(T, device=x.device))
        h = tok + pos

        for block in self.blocks:
            h = block(h)

        h = self.final_norm(h)
        return self.lm_head(h)


def build_model(vocab_size, block_size):
    d_model = 768
    n_layers = 8
    n_heads = 8
    d_ff = 3072

    return LegalDecoderLM(
        vocab_size=vocab_size,
        block_size=block_size,
        d_model=d_model,
        n_layers=n_layers,
        n_heads=n_heads,
        d_ff=d_ff,
    )
'''

_ARCH2_SRC = r'''class MultiHeadAttention(nn.Module):

    def __init__(
        self,
        d_model,
        n_heads,
        dropout=0.1
    ):
        super().__init__()

        assert d_model % n_heads == 0

        self.d_model = d_model
        self.n_heads = n_heads
        self.d_head = d_model // n_heads

        self.q_proj = nn.Linear(
            d_model,
            d_model
        )

        self.k_proj = nn.Linear(
            d_model,
            d_model
        )

        self.v_proj = nn.Linear(
            d_model,
            d_model
        )

        self.out_proj = nn.Linear(
            d_model,
            d_model
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(
        self,
        query,
        key_value,
        mask=None,
        is_causal=False
    ):

        B, Tq, _ = query.shape
        _, Tk, _ = key_value.shape

        Q = self.q_proj(query)
        K = self.k_proj(key_value)
        V = self.v_proj(key_value)

        Q = Q.view(
            B,
            Tq,
            self.n_heads,
            self.d_head
        ).transpose(1, 2)

        K = K.view(
            B,
            Tk,
            self.n_heads,
            self.d_head
        ).transpose(1, 2)

        V = V.view(
            B,
            Tk,
            self.n_heads,
            self.d_head
        ).transpose(1, 2)

        dropout_p = (
            self.dropout.p
            if self.training
            else 0.0
        )

        if is_causal:

            # Native causal kernel.
            context = F.scaled_dot_product_attention(
                Q,
                K,
                V,
                attn_mask=None,
                dropout_p=dropout_p,
                is_causal=True,
            )

        else:

            if mask is not None:
                mask = mask.to(
                    device=Q.device,
                    dtype=Q.dtype
                )

            context = F.scaled_dot_product_attention(
                Q,
                K,
                V,
                attn_mask=mask,
                dropout_p=dropout_p,
                is_causal=False,
            )

        context = (
            context
            .transpose(1, 2)
            .contiguous()
            .view(
                B,
                Tq,
                self.d_model
            )
        )

        return self.out_proj(context)


# ============================================================
# 5. SINUSOIDAL POSITIONAL ENCODING
# ============================================================

def make_sinusoidal_table(
    max_len,
    d_model
):

    position = torch.arange(
        max_len,
        dtype=torch.float32
    ).unsqueeze(1)

    k = torch.arange(
        0,
        d_model,
        2,
        dtype=torch.float32
    )

    omega = 1.0 / (
        10000 ** (k / d_model)
    )

    table = torch.zeros(
        max_len,
        d_model,
        dtype=torch.float32
    )

    table[:, 0::2] = torch.sin(
        position * omega
    )

    table[:, 1::2] = torch.cos(
        position * omega
    )

    return table


# ============================================================
# 6. FEED-FORWARD
# ============================================================

class FeedForward(nn.Module):

    def __init__(
        self,
        d_model,
        d_ff,
        dropout=0.1
    ):
        super().__init__()

        self.linear1 = nn.Linear(
            d_model,
            d_ff
        )

        self.activation = nn.GELU()

        self.linear2 = nn.Linear(
            d_ff,
            d_model
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(self, x):

        x = self.linear1(x)
        x = self.activation(x)
        x = self.linear2(x)
        x = self.dropout(x)

        return x


# ============================================================
# 7. PRE-LN ENCODER LAYER
# ============================================================

class TransformerEncoderLayerPreLN(
    nn.Module
):

    def __init__(
        self,
        d_model,
        n_heads,
        d_ff,
        dropout=0.1
    ):
        super().__init__()

        self.norm1 = nn.LayerNorm(
            d_model
        )

        self.self_attn = MultiHeadAttention(
            d_model=d_model,
            n_heads=n_heads,
            dropout=dropout
        )

        self.norm2 = nn.LayerNorm(
            d_model
        )

        self.ffn = FeedForward(
            d_model=d_model,
            d_ff=d_ff,
            dropout=dropout
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(self, x):

        # Pre-LN:
        # x + Attention(LN(x))

        nx = self.norm1(x)

        x = x + self.dropout(
            self.self_attn(
                nx,
                nx
            )
        )

        # Pre-LN:
        # x + FFN(LN(x))

        x = x + self.ffn(
            self.norm2(x)
        )

        return x


# ============================================================
# 8. PRE-LN ENCODER
# ============================================================

class TransformerEncoderPreLN(
    nn.Module
):

    def __init__(
        self,
        d_model=768,
        n_heads=8,
        d_ff=3072,
        n_layers=6,
        dropout=0.1
    ):
        super().__init__()

        self.layers = nn.ModuleList([
            TransformerEncoderLayerPreLN(
                d_model=d_model,
                n_heads=n_heads,
                d_ff=d_ff,
                dropout=dropout
            )
            for _ in range(n_layers)
        ])

    def forward(self, x):

        for layer in self.layers:
            x = layer(x)

        return x


# ============================================================
# 9. PRE-LN DECODER LAYER
# ============================================================

class TransformerDecoderLayerPreLN(
    nn.Module
):

    def __init__(
        self,
        d_model,
        n_heads,
        d_ff,
        dropout=0.1
    ):
        super().__init__()

        self.norm1 = nn.LayerNorm(
            d_model
        )

        self.self_attn = MultiHeadAttention(
            d_model=d_model,
            n_heads=n_heads,
            dropout=dropout
        )

        self.norm2 = nn.LayerNorm(
            d_model
        )

        self.cross_attn = MultiHeadAttention(
            d_model=d_model,
            n_heads=n_heads,
            dropout=dropout
        )

        self.norm3 = nn.LayerNorm(
            d_model
        )

        self.ffn = FeedForward(
            d_model=d_model,
            d_ff=d_ff,
            dropout=dropout
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(
        self,
        x,
        memory
    ):

        # ----------------------------------------------------
        # 1. Causal self-attention
        # ----------------------------------------------------

        nx = self.norm1(x)

        x = x + self.dropout(
            self.self_attn(
                nx,
                nx,
                is_causal=True
            )
        )

        # ----------------------------------------------------
        # 2. Cross-attention
        # Q = decoder
        # K/V = encoder
        # ----------------------------------------------------

        x = x + self.dropout(
            self.cross_attn(
                self.norm2(x),
                memory
            )
        )

        # ----------------------------------------------------
        # 3. FFN
        # ----------------------------------------------------

        x = x + self.ffn(
            self.norm3(x)
        )

        return x


# ============================================================
# 10. PRE-LN DECODER
# ============================================================

class TransformerDecoderPreLN(
    nn.Module
):

    def __init__(
        self,
        d_model=768,
        n_heads=8,
        d_ff=3072,
        n_layers=6,
        max_len=256,
        dropout=0.1
    ):
        super().__init__()

        self.layers = nn.ModuleList([
            TransformerDecoderLayerPreLN(
                d_model=d_model,
                n_heads=n_heads,
                d_ff=d_ff,
                dropout=dropout
            )
            for _ in range(n_layers)
        ])

        # Kept for diagnostics/check compatibility.
        self.register_buffer(
            "causal_mask",
            torch.triu(
                torch.full(
                    (max_len, max_len),
                    float("-inf")
                ),
                diagonal=1
            ),
            persistent=False
        )

    def forward(
        self,
        x,
        memory
    ):

        for layer in self.layers:

            x = layer(
                x,
                memory
            )

        return x


# ============================================================
# 11. COMPLETE PRE-LN MODEL
# ============================================================

class TransformerModelPreLN(
    nn.Module
):

    def __init__(
        self,
        vocab_size,
        max_src_len,
        max_tgt_len,
        d_model=768,
        n_heads=8,
        d_ff=3072,
        n_layers=6,
        dropout=0.1
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.d_model = d_model
        self.max_src_len = max_src_len
        self.max_tgt_len = max_tgt_len

        # Shared embedding.
        self.token_embedding = nn.Embedding(
            vocab_size,
            d_model
        )

        # Fixed source positions.
        self.register_buffer(
            "src_pos_encoding",
            make_sinusoidal_table(
                max_src_len,
                d_model
            ),
            persistent=False
        )

        # Fixed target positions.
        self.register_buffer(
            "tgt_pos_encoding",
            make_sinusoidal_table(
                max_tgt_len,
                d_model
            ),
            persistent=False
        )

        self.dropout = nn.Dropout(
            dropout
        )

        self.encoder = TransformerEncoderPreLN(
            d_model=d_model,
            n_heads=n_heads,
            d_ff=d_ff,
            n_layers=n_layers,
            dropout=dropout
        )

        self.decoder = TransformerDecoderPreLN(
            d_model=d_model,
            n_heads=n_heads,
            d_ff=d_ff,
            n_layers=n_layers,
            max_len=max_tgt_len,
            dropout=dropout
        )

        # Tied output head.
        self.output_projection = nn.Linear(
            d_model,
            vocab_size,
            bias=False
        )

        self.output_projection.weight = (
            self.token_embedding.weight
        )

        self._init_weights()

    def _init_weights(self):

        nn.init.normal_(
            self.token_embedding.weight,
            mean=0.0,
            std=0.02
        )

        for module in self.modules():

            if isinstance(
                module,
                nn.Linear
            ):

                if (
                    module.weight
                    is not self.token_embedding.weight
                ):

                    nn.init.xavier_uniform_(
                        module.weight
                    )

                if module.bias is not None:
                    nn.init.zeros_(
                        module.bias
                    )

            elif isinstance(
                module,
                nn.LayerNorm
            ):

                nn.init.ones_(
                    module.weight
                )

                nn.init.zeros_(
                    module.bias
                )

    def encode(self, src):

        _, T = src.shape

        assert T <= self.max_src_len

        x = self.token_embedding(src)

        x = x * math.sqrt(
            self.d_model
        )

        x = (
            x
            + self.src_pos_encoding[
                :T
            ].unsqueeze(0)
        )

        x = self.dropout(x)

        return self.encoder(x)

    def decode(
        self,
        tgt_in,
        memory
    ):

        _, T = tgt_in.shape

        assert T <= self.max_tgt_len

        x = self.token_embedding(
            tgt_in
        )

        x = x * math.sqrt(
            self.d_model
        )

        x = (
            x
            + self.tgt_pos_encoding[
                :T
            ].unsqueeze(0)
        )

        x = self.dropout(x)

        x = self.decoder(
            x,
            memory
        )

        return self.output_projection(x)

    def forward(
        self,
        src,
        tgt_in
    ):

        memory = self.encode(src)

        return self.decode(
            tgt_in,
            memory
        )


# ============================================================
# 12. BUILD MODEL
# ============================================================

def build_model(
    vocab_size,
    src_len,
    tgt_len
):

    assert src_len + tgt_len == 1024

    return TransformerModelPreLN(
        vocab_size=vocab_size,
        max_src_len=src_len,
        max_tgt_len=tgt_len,
        d_model=768,
        n_heads=8,
        d_ff=3072,
        n_layers=6,
        dropout=0.1
    )
'''

_ARCH3_SRC = r'''def sinusoidal_pos_emb(T, d_model, device):
    """
    Fixed sinusoidal position encoding.
    Shape: (1, T, d_model). No learned parameters.
    """
    pos = torch.arange(T, device=device, dtype=torch.float32).unsqueeze(1)

    div = torch.exp(
        torch.arange(0, d_model, 2, device=device, dtype=torch.float32)
        * (-math.log(10000.0) / d_model)
    )

    pe = torch.zeros(T, d_model, device=device, dtype=torch.float32)
    pe[:, 0::2] = torch.sin(pos * div)
    pe[:, 1::2] = torch.cos(pos * div)

    return pe.unsqueeze(0)  # (1, T, d_model)


class SlidingWindowCausalSelfAttention(nn.Module):
    def __init__(self, d_model, n_heads, block_size, window_size):
        super().__init__()

        assert d_model % n_heads == 0

        self.d_model = d_model
        self.n_heads = n_heads
        self.head_dim = d_model // n_heads
        self.window_size = window_size

        self.q_proj = nn.Linear(d_model, d_model)
        self.k_proj = nn.Linear(d_model, d_model)
        self.v_proj = nn.Linear(d_model, d_model)
        self.out_proj = nn.Linear(d_model, d_model)

        # Mask rule:
        # token i may attend to token j iff j <= i and i - j < W.
        #
        # For torch.nn.functional.scaled_dot_product_attention,
        # a boolean mask means True = allowed.
        idx = torch.arange(block_size)

        causal = idx[None, :] <= idx[:, None]          # j <= i
        window = (idx[:, None] - idx[None, :]) < window_size  # i - j < W

        mask = causal & window

        self.register_buffer(
            "swa_mask",
            mask.view(1, 1, block_size, block_size),
            persistent=False,
        )

    def forward(self, x):
        B, T, C = x.shape

        q = self.q_proj(x)
        k = self.k_proj(x)
        v = self.v_proj(x)

        q = q.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        k = k.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)
        v = v.view(B, T, self.n_heads, self.head_dim).transpose(1, 2)

        # Slice the precomputed mask for shorter inputs.
        attn_mask = self.swa_mask[:, :, :T, :T]

        # Do NOT use is_causal=True together with an explicit mask.
        # The banded mask already includes causality.
        y = F.scaled_dot_product_attention(
            q,
            k,
            v,
            attn_mask=attn_mask,
            dropout_p=0.0,
            is_causal=False,
        )

        y = y.transpose(1, 2).contiguous().view(B, T, C)

        return self.out_proj(y)


class FeedForward(nn.Module):
    def __init__(self, d_model, d_ff):
        super().__init__()

        self.fc1 = nn.Linear(d_model, d_ff)
        self.fc2 = nn.Linear(d_ff, d_model)

    def forward(self, x):
        return self.fc2(F.gelu(self.fc1(x)))


class TransformerBlock(nn.Module):
    def __init__(self, d_model, n_heads, d_ff, block_size, window_size):
        super().__init__()

        self.norm1 = nn.LayerNorm(d_model)

        self.attn = SlidingWindowCausalSelfAttention(
            d_model=d_model,
            n_heads=n_heads,
            block_size=block_size,
            window_size=window_size,
        )

        self.norm2 = nn.LayerNorm(d_model)

        self.mlp = FeedForward(d_model, d_ff)

    def forward(self, x):
        # Pre-LN residual block.
        x = x + self.attn(self.norm1(x))
        x = x + self.mlp(self.norm2(x))
        return x


class LegalDecoderLM(nn.Module):
    def __init__(
        self,
        vocab_size,
        block_size,
        d_model,
        n_layers,
        n_heads,
        d_ff,
        window_size,
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.block_size = block_size
        self.d_model = d_model
        self.n_layers = n_layers

        self.token_emb = nn.Embedding(vocab_size, d_model)

        # Scale token embeddings by sqrt(d_model) so they are comparable
        # in magnitude to the sinusoidal position table.
        self.embed_scale = math.sqrt(d_model)

        # Precompute sinusoidal positions once. Not a parameter.
        pos = sinusoidal_pos_emb(block_size, d_model, device="cpu")
        self.register_buffer("pos_emb", pos, persistent=False)

        self.blocks = nn.ModuleList([
            TransformerBlock(
                d_model=d_model,
                n_heads=n_heads,
                d_ff=d_ff,
                block_size=block_size,
                window_size=window_size,
            )
            for _ in range(n_layers)
        ])

        self.final_norm = nn.LayerNorm(d_model)

        self.lm_head = nn.Linear(d_model, vocab_size, bias=False)

        # Tied embedding and output head.
        self.lm_head.weight = self.token_emb.weight

        self.apply(self._init_weights)

        # GPT-style residual scaling for deep pre-LN stacks.
        for block in self.blocks:
            block.attn.out_proj.weight.data.mul_(1.0 / math.sqrt(2 * self.n_layers))
            block.mlp.fc2.weight.data.mul_(1.0 / math.sqrt(2 * self.n_layers))

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)
            if module.bias is not None:
                nn.init.zeros_(module.bias)

        elif isinstance(module, nn.Embedding):
            nn.init.normal_(module.weight, mean=0.0, std=0.02)

        elif isinstance(module, nn.LayerNorm):
            nn.init.ones_(module.weight)
            nn.init.zeros_(module.bias)

    def forward(self, x):
        B, T = x.shape

        assert T <= self.block_size, (
            f"Sequence length {T} exceeds block_size {self.block_size}"
        )

        tok = self.token_emb(x) * self.embed_scale

        pos = self.pos_emb[:, :T, :].to(tok.dtype)

        h = tok + pos

        for block in self.blocks:
            h = block(h)

        h = self.final_norm(h)

        return self.lm_head(h)


def build_model(vocab_size, block_size, window_size=256):
    """
    Architecture 3 size:
      d_model = 1024
      n_layers = 16
      n_heads = 16
      d_ff = 4096
      W = 256 by default

    Parameter count with vocab_size=40000:
      242,501,632 ≈ 242.5M
    """
    d_model = 1024
    n_layers = 16
    n_heads = 16
    d_ff = 4096

    return LegalDecoderLM(
        vocab_size=vocab_size,
        block_size=block_size,
        d_model=d_model,
        n_layers=n_layers,
        n_heads=n_heads,
        d_ff=d_ff,
        window_size=window_size,
    )
'''


def _namespace(src, name):
    ns = {"math": math, "torch": torch, "nn": nn, "F": F, "__name__": f"phase1_{name}"}
    exec(compile(src, f"<phase1 {name} definition>", "exec"), ns)
    return ns


# ======================= loading =======================
def _strip_compile_prefix(sd):
    return {(k[len("_orig_mod."):] if k.startswith("_orig_mod.") else k): v for k, v in sd.items()}


def load_backbone(name, ckpt_path=None, device="cuda"):
    """Returns SimpleNamespace(model, hidden, max_len, d_model, meta). Weights are loaded strict=True."""
    ckpt_path = ckpt_path or os.environ.get(f"CKPT_{name.upper()}") or DEFAULT_CKPTS[name]
    if name == "arch1":
        ns = _namespace(_ARCH1_SRC, name)
        model, max_len, d_model = ns["build_model"](VOCAB, 1024), 1024, 768

        def hidden(ids):
            B, T = ids.shape
            assert T <= max_len
            h = model.token_emb(ids) + model.pos_emb(torch.arange(T, device=ids.device))
            for block in model.blocks:
                h = block(h)
            return model.final_norm(h)
    elif name == "arch3":
        ns = _namespace(_ARCH3_SRC, name)
        model, max_len, d_model = ns["build_model"](VOCAB, 1024, 256), 1024, 1024

        def hidden(ids):
            B, T = ids.shape
            assert T <= max_len
            h = model.token_emb(ids) * model.embed_scale + model.pos_emb[:, :T, :].to(model.token_emb.weight.dtype)
            for block in model.blocks:
                h = block(h)
            return model.final_norm(h)
    elif name == "arch2":
        ns = _namespace(_ARCH2_SRC, name)
        model, max_len, d_model = ns["build_model"](VOCAB, 768, 256), 768, 768

        def hidden(ids):
            return model.encode(ids)               # raw encoder output; there is no final norm
    else:
        raise ValueError("name must be arch1, arch2 or arch3")

    ck = torch.load(ckpt_path, map_location="cpu", weights_only=False)
    state = ck["model"] if isinstance(ck, dict) and "model" in ck else ck
    model.load_state_dict(_strip_compile_prefix(state), strict=True)
    n_params = sum(p.numel() for p in model.parameters())
    meta = {"name": name, "ckpt": ckpt_path, "n_params": n_params,
            "step": ck.get("step") if isinstance(ck, dict) else None,
            "best": ck.get("best") if isinstance(ck, dict) else None}
    model = model.to(device).eval()
    return SimpleNamespace(model=model, hidden=hidden, max_len=max_len, d_model=d_model, meta=meta)


# ======================= self-test =======================
@torch.inference_mode()
def selftest(name, ckpt_path=None, device="cuda"):
    print("\n" + "=" * 72 + f"\n{name}\n" + "=" * 72)
    bb = load_backbone(name, ckpt_path, device)
    m, meta = bb.model, bb.meta
    print(f"checkpoint: {meta['ckpt']}\n  step={meta['step']}  best={meta['best']}")
    print(f"  parameters: {meta['n_params']:,}  (expected {EXPECTED_PARAMS[name]:,})")
    assert meta["n_params"] == EXPECTED_PARAMS[name], "parameter count differs from the Phase-1 model"
    print("  strict load: PASS | parameter count: PASS")

    g = torch.Generator(device="cpu").manual_seed(0)
    T = 512 if name != "arch2" else 512
    x = torch.randint(0, VOCAB, (2, T), generator=g).to(device)
    h = bb.hidden(x).float()
    assert h.shape == (2, T, bb.d_model) and torch.isfinite(h).all()
    print(f"  hidden shape {tuple(h.shape)}: PASS")

    if name in ("arch1", "arch3"):
        diff = (m.lm_head(bb.hidden(x)).float() - m(x).float()).abs().max().item()
        print(f"  parity  lm_head(hidden(x)) vs model(x): max|diff| = {diff:.2e}")
        assert diff < 1e-3, "hidden() does not reproduce the model's own forward pass"
        t = T // 2
        x2 = x.clone()
        x2[:, t:] = torch.randint(0, VOCAB, x2[:, t:].shape, generator=g).to(device)
        leak = (bb.hidden(x).float()[:, :t] - bb.hidden(x2).float()[:, :t]).abs().max().item()
        print(f"  causality: changing tokens >= {t} moves earlier states by {leak:.2e} (must be ~0)")
        assert leak < 1e-3
    else:
        x2 = x.clone()
        x2[:, -1] = (x2[:, -1] + 1) % VOCAB
        early = (bb.hidden(x).float()[:, 0] - bb.hidden(x2).float()[:, 0]).abs().max().item()
        print(f"  bidirectional encoder: changing ONLY the last token moves position-0 state by {early:.2e} (expected > 0)")
        assert early > 0
        print("  (no padding mask exists: padded positions are attended to like real tokens)")

    n = h.norm(dim=-1)
    print(f"  activation scale: per-token L2 norm mean {n.mean():.2f}, std {n.std():.2f}, max {n.max():.2f} | "
          f"max |value| {h.abs().max():.2f} | per-dimension std (median) {h.std(dim=(0, 1)).median():.3f}")
    return bb


def selftest_all(device="cuda"):
    for name in ("arch1", "arch2", "arch3"):
        try:
            selftest(name, None, device)
            torch.cuda.empty_cache()
        except Exception as e:                                   # keep going; show which one failed
            print(f"\nFAILED {name}: {type(e).__name__}: {e}")




# Find the original Phase 1 Arch3 checkpoint and tokenizer from attached Kaggle inputs.
INPUT_ROOT = Path("/kaggle/input")

arch3_default = Path(os.environ.get("CKPT_ARCH3", DEFAULT_CKPTS["arch3"]))
if arch3_default.is_file():
    ARCH3_CKPT = arch3_default
else:
    candidates = [p for p in INPUT_ROOT.rglob("best.pt") if "arch3" in str(p).lower()]
    candidates = sorted(candidates, key=lambda p: ("arch3-model" not in str(p).lower(), len(str(p))))
    if not candidates:
        raise FileNotFoundError(
            "Arch3 Phase 1 best.pt was not found. Attach the Phase 1 Arch3 model/dataset to this Kaggle notebook. "
            f"Expected default: {DEFAULT_CKPTS['arch3']}"
        )
    ARCH3_CKPT = candidates[0]

preferred_tokenizer = Path("/kaggle/input/datasets/achiver07/phase1-tokens-v40k-clean1/tokenizer.json")
if preferred_tokenizer.is_file():
    TOKENIZER_PATH = preferred_tokenizer
else:
    toks = sorted(INPUT_ROOT.rglob("tokenizer.json"), key=lambda p: ("phase1-tokens-v40k-clean1" not in str(p).lower(), len(str(p))))
    if not toks:
        raise FileNotFoundError("Phase 1 tokenizer.json not found under /kaggle/input; attach the Phase 1 tokenizer dataset.")
    TOKENIZER_PATH = toks[0]

if not all(p.is_file() for p in (TRAIN_PATH, DEV_PATH, TEST_PATH)):
    print("Phase 4 files not found in /kaggle/working; attempting Hugging Face dataset download...")
    try:
        from kaggle_secrets import UserSecretsClient
        from huggingface_hub import snapshot_download
        hf_token = UserSecretsClient().get_secret("HF_TOKEN")
        snapshot_download(
            repo_id="Exploration-Lab/CS-781-Capstone",
            repo_type="dataset",
            token=hf_token,
            allow_patterns=["Phase_4_CJPE/**"],
            local_dir="/kaggle/working/Phase_4_CJPE",
        )
    except Exception as exc:
        raise RuntimeError(
            "CJPE files are missing and automatic download failed. Check Kaggle Internet, "
            "the HF_TOKEN secret, and repository access."
        ) from exc

for p in (TRAIN_PATH, DEV_PATH, TEST_PATH):
    if not p.is_file():
        raise FileNotFoundError(f"CJPE file missing after download: {p}")

print("Arch3 checkpoint:", ARCH3_CKPT)
print("Tokenizer:", TOKENIZER_PATH)
print("Expected Arch3 parameter count:", f"{EXPECTED_PARAMS['arch3']:,}")

# Load train/dev labels, merge both labelled splits into the final training corpus,
# and deliberately discard every field except id/text from test examples.
def load_json(path):
    with open(path, "r", encoding="utf-8") as f:
        return json.load(f)

train_split_records = load_json(TRAIN_PATH)
dev_split_records = load_json(DEV_PATH)
test_raw = load_json(TEST_PATH)
test_records = [{"id": row["id"], "text": row.get("text", "")} for row in test_raw]
del test_raw

for name, records in [("train", train_split_records), ("dev", dev_split_records)]:
    assert records, f"{name} split is empty"
    assert all("id" in r and "text" in r for r in records), f"Malformed {name} example"
    assert all(r.get("label") in (0, 1) for r in records), f"Invalid/missing labels in {name}"

assert test_records, "test split is empty"
assert all("id" in r and "text" in r for r in test_records), "Malformed test example"

train_ids_original = [str(r["id"]) for r in train_split_records]
dev_ids_original = [str(r["id"]) for r in dev_split_records]
test_ids = [str(r["id"]) for r in test_records]

# Test IDs must be unique because submission output requires one prediction per test ID.
assert len(set(test_ids)) == len(test_ids), "Duplicate IDs in test split: submission must contain one prediction per test ID"

# Do NOT reject repeated IDs in train/dev. The assignment's full-corpus run uses every
# labelled row, and repeated case IDs can occur in the supplied combined data. Training
# consumes rows (text + label), not IDs as dictionary keys, so keep all examples.
train_dup_rows = len(train_ids_original) - len(set(train_ids_original))
dev_dup_rows = len(dev_ids_original) - len(set(dev_ids_original))
train_dev_id_overlap = len(set(train_ids_original) & set(dev_ids_original))
print(f"Repeated-ID rows retained in train: {train_dup_rows:,}")
print(f"Repeated-ID rows retained in dev:   {dev_dup_rows:,}")
print(f"Distinct IDs appearing in both train/dev: {train_dev_id_overlap:,}")

SOURCE_TRAIN_COUNT = len(train_split_records)
SOURCE_DEV_COUNT = len(dev_split_records)
corpus_records = train_split_records + dev_split_records

# Optional smoke-test switch; leave None for the required full-corpus run.
if MAX_TRAIN_DOCS is not None:
    rng = np.random.default_rng(SEED)
    n_keep = min(int(MAX_TRAIN_DOCS), len(corpus_records))
    chosen = np.sort(rng.choice(len(corpus_records), size=n_keep, replace=False))
    corpus_records = [corpus_records[i] for i in chosen]

train_records = corpus_records
y_train = np.asarray([int(r["label"]) for r in train_records], dtype=np.int64)
train_ids = [str(r["id"]) for r in train_records]
train_test_id_overlap = len(set(train_ids) & set(test_ids))
if train_test_id_overlap:
    # Report rather than discard any labelled rows; user requested full-corpus training.
    # No test labels are loaded or used. Investigate this if the competition requires
    # disjoint IDs, because overlapping IDs can indicate a split/data-integrity issue.
    print(f"WARNING: {train_test_id_overlap:,} distinct case IDs occur in both labelled data and test. All labelled rows are retained; test labels are not used.")

print(f"Original train split: {SOURCE_TRAIN_COUNT:,}")
print(f"Development examples merged into training: {SOURCE_DEV_COUNT:,}")
print(f"Full labelled training corpus: {len(train_records):,}")
print(f"Test examples (id/text only): {len(test_records):,}")
print("Combined corpus labels:", Counter(y_train.tolist()))
print("Evaluation: DISABLED; development labels are part of training.")
print("Test labels intentionally not loaded or used.")

# Load the project's custom 40k tokenizer.
tokenizer = Tokenizer.from_file(str(TOKENIZER_PATH))


BOS_ID, EOS_ID, PAD_ID = 1, 2, 0
N_ZONES = 4


def select_token_zones(full_token_ids, zone_tokens):
    """Select four token windows across the full case, including the true ending.

    Short cases (<= 4 windows) are split into four non-overlapping contiguous
    sections so every token is represented. Long cases use four windows near
    the beginning, 1/3, 2/3, and the final token. All slicing happens in token
    space, not character space, and the tokenizer is never truncated first.
    """
    ids = list(map(int, full_token_ids))
    n = len(ids)
    if n == 0:
        return [[], [], [], []]

    if n <= N_ZONES * zone_tokens:
        boundaries = np.linspace(0, n, N_ZONES + 1, dtype=int)
        return [ids[boundaries[i]:boundaries[i + 1]] for i in range(N_ZONES)]

    last_start = n - zone_tokens
    starts = [
        0,
        int(round(last_start / 3)),
        int(round(2 * last_start / 3)),
        last_start,
    ]
    # Keep windows sorted and bounded; final window always reaches document end.
    starts = [max(0, min(s, last_start)) for s in starts]
    starts[-1] = last_start
    return [ids[s:s + zone_tokens] for s in starts]


def encode_records(records, split_name, batch_docs=32):
    """Tokenize complete documents and choose token-based zones (no character truncation)."""
    n = len(records)
    content_len = MAX_ZONE_LEN - 2
    token_ids = np.zeros((n, N_ZONES, MAX_ZONE_LEN), dtype=np.int32)
    lengths = np.zeros((n, N_ZONES), dtype=np.int16)

    # Explicitly disable truncation so that the real final tokens remain available.
    tokenizer.no_truncation()
    for start in tqdm(range(0, n, batch_docs), desc=f"Tokenizing {split_name}"):
        batch = records[start:start + batch_docs]
        texts = [str(row.get("text") or "") for row in batch]
        encodings = tokenizer.encode_batch(texts, add_special_tokens=False)
        for offset, enc in enumerate(encodings):
            doc_i = start + offset
            zones = select_token_zones(enc.ids, content_len)
            for zone_i, zone in enumerate(zones):
                content = [int(t) for t in zone[:content_len]]
                seq = [BOS_ID] + content + [EOS_ID]
                seq_len = min(len(seq), MAX_ZONE_LEN)
                token_ids[doc_i, zone_i, :seq_len] = seq[:seq_len]
                lengths[doc_i, zone_i] = seq_len

    assert token_ids.min() >= 0 and token_ids.max() < VOCAB, (
        "Tokenizer emitted a token ID outside the Phase 1 vocabulary."
    )
    return token_ids, lengths


# Final fit only: train and development labels both belong to the training corpus.
# Test input is id/text only; test labels are not used.
X_train, L_train = encode_records(train_records, "full labelled corpus", batch_docs=32)
X_test, L_test = encode_records(test_records, "test inputs (text only)", batch_docs=32)

print("Encoded shapes:", X_train.shape, X_test.shape)
print("Mean zone lengths (training corpus):", np.round(L_train.mean(axis=0), 1).tolist())
print("Token selection: full-tokenizer output; final window reaches document end.")

print("PEFT version:", getattr(peft, "__version__", "unknown"))

# Load only the Phase 1 Arch3 checkpoint. Strict load and expected parameter count are checked by the helper.
bb3 = load_backbone("arch3", ckpt_path=str(ARCH3_CKPT), device=str(DEVICE))
assert bb3.meta["n_params"] == EXPECTED_PARAMS["arch3"], "Loaded model is not the expected Phase 1 Arch3 checkpoint."
print("Loaded:", bb3.meta)

# Verify the exact eligible modules before adapter injection.
TARGET_SUFFIXES = ("q_proj", "k_proj", "v_proj", "out_proj", "fc1", "fc2")
target_names = [name for name, module in bb3.model.named_modules()
                if isinstance(module, nn.Linear) and name.split(".")[-1] in TARGET_SUFFIXES]
print("Linear modules selected for AdaLoRA:", len(target_names))
print("Examples:", target_names[:12])
assert len(target_names) == 16 * 6, f"Expected 96 Arch3 attention/MLP projections, found {len(target_names)}"

train_loader = DataLoader(
    TensorDataset(torch.from_numpy(X_train).long(), torch.from_numpy(L_train).long(), torch.from_numpy(y_train).long()),
    batch_size=BATCH_DOCS, shuffle=True, num_workers=0, pin_memory=True,
)
steps_per_epoch = len(train_loader)
total_steps = EPOCHS * steps_per_epoch
assert total_steps >= 4, f"Too few optimization steps for AdaLoRA schedule: {total_steps}"
TINIT = max(1, int(total_steps * 0.05))
TFINAL = max(1, int(total_steps * 0.10))
if TINIT >= total_steps - TFINAL:
    TINIT, TFINAL = 0, max(1, total_steps // 5)

adalora_config = AdaLoraConfig(
    init_r=8,
    target_r=4,
    tinit=TINIT,
    tfinal=TFINAL,
    deltaT=25,
    beta1=0.85,
    beta2=0.85,
    lora_alpha=16,
    lora_dropout=0.05,
    orth_reg_weight=0.1,
    total_step=total_steps,
    target_modules=list(TARGET_SUFFIXES),
    task_type=None,
    bias="none",
)

# Adapter injection modifies Arch3's target linear projections in-place while preserving its Phase 1 base weights.
adalora_model = get_peft_model(bb3.model, adalora_config)
adalora_model.train()
adalora_model.print_trainable_parameters()

# Custom classification head: attention pooling across four zone representations + tail/ratio evidence.
class CJPEZoneClassifier(nn.Module):
    def __init__(self, d_model=1024, dropout=0.25):
        super().__init__()
        self.zone_score = nn.Sequential(nn.LayerNorm(d_model), nn.Linear(d_model, 128), nn.Tanh(), nn.Linear(128, 1))
        self.classifier = nn.Sequential(
            nn.LayerNorm(d_model * 4),
            nn.Dropout(dropout),
            nn.Linear(d_model * 4, 512),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(512, 2),
        )
    def forward(self, zone_reps):
        # zone_reps: [batch, 4, d_model]
        weights = torch.softmax(self.zone_score(zone_reps).squeeze(-1), dim=1)
        pooled = torch.sum(zone_reps * weights.unsqueeze(-1), dim=1)
        ratio = zone_reps[:, 2]
        holding = zone_reps[:, 3]
        fused = torch.cat([pooled, ratio, holding, holding - ratio], dim=-1)
        return self.classifier(fused), weights

classifier = CJPEZoneClassifier(d_model=bb3.d_model).to(DEVICE)

# Use gradient checkpointing for the 16-layer frozen-base / trainable-adapter path.
def arch3_hidden(ids):
    model = bb3.model
    B, T = ids.shape
    if T > bb3.max_len:
        raise ValueError(f"Input length {T} exceeds Arch3 max length {bb3.max_len}")
    h = model.token_emb(ids) * model.embed_scale + model.pos_emb[:, :T, :].to(model.token_emb.weight.dtype)
    for block in model.blocks:
        if USE_GRAD_CHECKPOINTING and model.training and torch.is_grad_enabled():
            h = checkpoint(block, h, use_reentrant=False)
        else:
            h = block(h)
    return model.final_norm(h)


def zone_representations(input_ids, lengths):
    # input_ids [B,4,L] -> one contextual vector per zone [B,4,D].
    B, Z, T = input_ids.shape
    flat_ids = input_ids.reshape(B * Z, T)
    flat_lengths = lengths.reshape(B * Z).clamp(min=1, max=T)
    h = arch3_hidden(flat_ids)
    positions = torch.arange(T, device=h.device).unsqueeze(0)
    mask = (positions < flat_lengths.unsqueeze(1)).unsqueeze(-1).to(h.dtype)
    mean_pool = (h * mask).sum(dim=1) / mask.sum(dim=1).clamp(min=1.0)
    row = torch.arange(B * Z, device=h.device)
    end_pool = h[row, flat_lengths - 1]
    return (0.5 * mean_pool + 0.5 * end_pool).reshape(B, Z, -1)

# Identity-preservation smoke test: newly initialized AdaLoRA should not materially perturb the base output.
probe = torch.from_numpy(X_train[:1]).long().to(DEVICE).reshape(-1, MAX_ZONE_LEN)
adalora_model.eval()
with torch.no_grad(), torch.amp.autocast("cuda", dtype=torch.float16):
    probe_hidden = arch3_hidden(probe)
assert probe_hidden.shape == (N_ZONES, MAX_ZONE_LEN, bb3.d_model)
print("AdaLoRA forward smoke test passed; hidden shape:", tuple(probe_hidden.shape))
del probe, probe_hidden
adalora_model.train()

adapter_params = [p for p in adalora_model.parameters() if p.requires_grad]
head_params = list(classifier.parameters())
trainable_params = adapter_params + head_params
n_base_total = int(bb3.meta["n_params"])  # original Phase 1 Arch3 count before adapters
n_trainable = sum(p.numel() for p in adapter_params) + sum(p.numel() for p in head_params)
optimizer = torch.optim.AdamW(
    [{"params": adapter_params, "lr": 8e-5}, {"params": head_params, "lr": 2e-4}],
    weight_decay=0.01,
)
counts = np.bincount(y_train, minlength=2).astype(np.float64)  # labels from combined train + dev corpus
class_weights = (len(y_train) / np.maximum(counts, 1.0)) ** 0.5
class_weights = class_weights / class_weights.mean()
criterion = nn.CrossEntropyLoss(weight=torch.tensor(class_weights, dtype=torch.float32, device=DEVICE), label_smoothing=0.02)
scaler = torch.amp.GradScaler("cuda", enabled=True)

print(f"AdaLoRA total steps={total_steps:,} | tinit={TINIT:,} | tfinal={TFINAL:,} | target rank=4")
print(f"Backbone params={n_base_total:,} | trainable (adapters + head)={n_trainable:,} ({100*n_trainable/(n_base_total+n_trainable):.3f}%)")

start_time = time.time()
epoch_losses = []

# Start a fresh adapter output folder; save final-epoch weights only (no dev-based selection).
if ADAPTER_DIR.exists():
    shutil.rmtree(ADAPTER_DIR)
ADAPTER_DIR.mkdir(parents=True, exist_ok=True)

for epoch in range(1, EPOCHS + 1):
    adalora_model.train(); classifier.train()
    running_loss = 0.0
    skipped_batches = 0
    pbar = tqdm(enumerate(train_loader, start=1), total=len(train_loader), desc=f"Full-corpus AdaLoRA epoch {epoch}/{EPOCHS}")
    for batch_idx, (ids, lens, labels) in pbar:
        ids = ids.to(DEVICE, non_blocking=True)
        lens = lens.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)
        optimizer.zero_grad(set_to_none=True)
        with torch.amp.autocast("cuda", dtype=torch.float16):
            z = zone_representations(ids, lens)
            logits, _ = classifier(z)
            loss = criterion(logits.float(), labels)
        if not torch.isfinite(loss):
            print(f"Non-finite loss at epoch={epoch}, batch={batch_idx}; batch skipped.")
            skipped_batches += 1
            optimizer.zero_grad(set_to_none=True)
            continue
        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(trainable_params, max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()
        global_step = (epoch - 1) * len(train_loader) + batch_idx - 1  # zero-based step index
        # REQUIRED: dynamic AdaLoRA rank allocation / importance update.
        adalora_model.base_model.update_and_allocate(global_step)
        running_loss += float(loss.detach().cpu())
        if batch_idx % 25 == 0:
            pbar.set_postfix(loss=f"{running_loss / max(1, batch_idx - skipped_batches):.4f}", step=global_step)

    processed_batches = max(1, len(train_loader) - skipped_batches)
    mean_train_loss = float(running_loss / processed_batches)
    epoch_losses.append({"epoch": int(epoch), "mean_train_loss": mean_train_loss, "skipped_batches": int(skipped_batches)})
    print(f"Epoch {epoch}: train_loss={mean_train_loss:.6f} | skipped_batches={skipped_batches}")

    # Save the latest/final epoch, with no development evaluation or checkpoint selection.
    adalora_model.save_pretrained(str(ADAPTER_DIR), safe_serialization=True)
    torch.save(classifier.state_dict(), HEAD_PATH)
    with open(OUT_DIR / "training_log.json", "w", encoding="utf-8") as f:
        json.dump({"epoch_losses": epoch_losses, "evaluation_performed": False}, f, indent=2)
    print("  Saved current AdaLoRA adapter + classifier head.")

train_seconds = time.time() - start_time
peak_gpu_gb = torch.cuda.max_memory_allocated() / 1024**3
print(f"Full-corpus training completed in {train_seconds/60:.1f} minutes | peak allocated GPU memory={peak_gpu_gb:.2f} GiB")
assert (ADAPTER_DIR / "adapter_config.json").is_file(), "AdaLoRA adapter checkpoint was not saved correctly."
assert HEAD_PATH.is_file(), "Classifier head checkpoint was not saved correctly."


# Free current training objects and reload the final full-corpus checkpoint on the original Phase 1 Arch3 weights.
# Delete only names that exist in this kernel (safe if a variable was renamed/cleared).
for _name in (
    "adalora_model", "bb3", "classifier", "optimizer", "scaler",
    "trainable_params", "adapter_params", "head_params"
):
    globals().pop(_name, None)
if torch.cuda.is_available():
    torch.cuda.empty_cache()
gc.collect()

bb3 = load_backbone("arch3", ckpt_path=str(ARCH3_CKPT), device=str(DEVICE))
adalora_model = PeftModel.from_pretrained(bb3.model, str(ADAPTER_DIR), is_trainable=False)
adalora_model.eval()
classifier = CJPEZoneClassifier(d_model=bb3.d_model).to(DEVICE)
classifier.load_state_dict(torch.load(HEAD_PATH, map_location=DEVICE, weights_only=True))
classifier.eval()

# Inference uses normal forward passes; no test labels are available in test_records.
def arch3_hidden(ids):
    model = bb3.model
    B, T = ids.shape
    h = model.token_emb(ids) * model.embed_scale + model.pos_emb[:, :T, :].to(model.token_emb.weight.dtype)
    for block in model.blocks:
        h = block(h)
    return model.final_norm(h)

@torch.no_grad()
def predict_probs(ids_array, len_array, desc):
    ds = TensorDataset(torch.from_numpy(ids_array).long(), torch.from_numpy(len_array).long())
    loader = DataLoader(ds, batch_size=BATCH_DOCS, shuffle=False, num_workers=0, pin_memory=True)
    probs, attention = [], []
    for ids, lens in tqdm(loader, desc=desc):
        ids = ids.to(DEVICE, non_blocking=True)
        lens = lens.to(DEVICE, non_blocking=True)
        with torch.amp.autocast("cuda", dtype=torch.float16):
            z = zone_representations(ids, lens)
            logits, w = classifier(z)
        probs.extend(torch.softmax(logits.float(), dim=-1)[:, 1].cpu().numpy().tolist())
        attention.append(w.float().cpu().numpy())
    return np.asarray(probs), np.concatenate(attention, axis=0)

p_test, attn_test = predict_probs(X_test, L_test, "Test prediction (text only)")

# Fixed threshold: no development set or labels are used for threshold tuning.
PREDICTION_THRESHOLD = 0.5
preds = (p_test >= PREDICTION_THRESHOLD).astype(np.int64)

submission = [{"id": str(row["id"]), "pred": int(pred)} for row, pred in zip(test_records, preds)]
assert len(submission) == len(test_records)
assert [x["id"] for x in submission] == test_ids
assert len({x["id"] for x in submission}) == len(submission)
assert all(type(x["pred"]) is int and x["pred"] in (0, 1) for x in submission)

training_metadata = {
    "experiment": "Phase 4 Experiment 2 — Arch3 + AdaLoRA, improved token zones (256), full-corpus training, no evaluation",
    "architecture": "Phase 1 Arch3 (242.5M parameters)",
    "adaptation": "AdaLoRA (dynamic rank allocation)",
    "checkpoint": str(ARCH3_CKPT),
    "tokenizer": str(TOKENIZER_PATH),
    "original_train_examples": int(SOURCE_TRAIN_COUNT),
    "development_examples_merged_into_training": int(SOURCE_DEV_COUNT),
    "full_corpus_training_examples": int(len(train_records)),
    "test_examples": int(len(test_records)),
    "repeated_id_rows_in_original_train_retained": int(train_dup_rows),
    "repeated_id_rows_in_original_dev_retained": int(dev_dup_rows),
    "distinct_train_dev_id_overlap": int(train_dev_id_overlap),
    "distinct_train_test_id_overlap_warning": int(train_test_id_overlap),
    "epochs": int(EPOCHS),
    "batch_docs": int(BATCH_DOCS),
    "zone_length": int(MAX_ZONE_LEN),
    "target_modules": list(TARGET_SUFFIXES),
    "init_rank": 8,
    "target_rank": 4,
    "tinit": int(TINIT),
    "tfinal": int(TFINAL),
    "deltaT": 25,
    "prediction_threshold_fixed": float(PREDICTION_THRESHOLD),
    "epoch_training_log": epoch_losses,
    "training_seconds": float(train_seconds),
    "peak_gpu_memory_gib": float(peak_gpu_gb),
    "mean_zone_attention_test": attn_test.mean(axis=0).tolist(),
    "test_positive_predictions": int((preds == 1).sum()),
    "test_negative_predictions": int((preds == 0).sum()),
    "test_labels_used": False,
    "evaluation_performed": False,
    "phase1_base_parameters": int(n_base_total),
    "trainable_parameters_adalora_plus_head": int(n_trainable),
    "trainable_parameter_percent": float(100 * n_trainable / (n_base_total + n_trainable)),
}

with open(SUBMISSION_PATH, "w", encoding="utf-8") as f:
    json.dump(submission, f, ensure_ascii=False, indent=2)
with open(RESULTS_PATH, "w", encoding="utf-8") as f:
    json.dump(training_metadata, f, ensure_ascii=False, indent=2)
with zipfile.ZipFile(ZIP_PATH, "w", compression=zipfile.ZIP_DEFLATED) as zf:
    zf.write(SUBMISSION_PATH, arcname="prediction.json")

print("\n" + "=" * 72)
print("PHASE 4 EXPERIMENT 2 — ARCH3 + ADALORA FULL-CORPUS RUN")
print("=" * 72)
print("Evaluation: NOT PERFORMED (as requested)")
print(f"Original train examples:             {SOURCE_TRAIN_COUNT:,}")
print(f"Dev examples merged into training:   {SOURCE_DEV_COUNT:,}")
print(f"Total labelled training examples:    {len(train_records):,}")
print(f"Test examples predicted:             {len(submission):,}")
print(f"Fixed decision threshold:            {PREDICTION_THRESHOLD:.2f}")
print(f"Training time:                       {train_seconds/60:.1f} minutes")
print(f"Peak allocated GPU memory:           {peak_gpu_gb:.2f} GiB")
print("Test prediction counts:", Counter(int(x["pred"]) for x in submission))
print("First 3 predictions:", submission[:3])
print("Submission JSON:", SUBMISSION_PATH)
print("Submission ZIP:", ZIP_PATH)
print("Training metadata (no eval metrics):", RESULTS_PATH)

# Save PEFT's final per-module rank pattern/config after the dynamic rank allocator ran.
try:
    cfg = adalora_model.peft_config.get("default")
    rank_pattern = getattr(cfg, "rank_pattern", None) if cfg is not None else None
    rank_json = {}
    if rank_pattern:
        rank_json = {str(k): v for k, v in rank_pattern.items()}
    with open(OUT_DIR / "adalora_rank_pattern.json", "w", encoding="utf-8") as f:
        json.dump(rank_json, f, indent=2)
    print("Saved rank pattern entries:", len(rank_json))
    if rank_json:
        print("Sample rank assignments:", list(rank_json.items())[:12])
    else:
        print("This PEFT build did not expose a populated rank_pattern in the saved config; adapter_config.json is preserved.")
except Exception as exc:
    print("Could not serialize rank_pattern:", repr(exc))

# Format validation independent of any test labels.
with open(SUBMISSION_PATH, "r", encoding="utf-8") as f:
    check = json.load(f)
assert isinstance(check, list) and len(check) == len(test_ids)
assert [str(x["id"]) for x in check] == test_ids
assert all(set(x.keys()) == {"id", "pred"} for x in check)
assert all(type(x["pred"]) is int and x["pred"] in (0, 1) for x in check)
with zipfile.ZipFile(ZIP_PATH, "r") as zf:
    assert zf.namelist() == ["prediction.json"]
print("Submission format validation: PASS (format validation only; no evaluation performed)")
print("Output folder contents:")
for p in sorted(OUT_DIR.rglob("*")):
    if p.is_file(): print(f"  {p} ({p.stat().st_size / 1024:.1f} KiB)")
```
